### Etapa 1: Importando e visualizando os dados
Vamos carregar a nossa tabela (dataset) com informações de saúde e visualizar as primeiras linhas para entender o que temos em mãos.

```python
import pandas as pd

# Carrega os dados do arquivo CSV
df = pd.read_csv("diabetes_pt_mat_apoio.csv")

# Exibe as 5 primeiras linhas da tabela
print("Visualizando os dados:")
display(df.head())
```

### Etapa 2: Importando 2 modelos de IA simples

Vamos importar dois modelos diferentes para aprender com nossos dados:

1. Árvore de Decisão: Cria regras simples (ex: se glicose > 120, então...).

2. Regressão Logística: Usa matemática para calcular a probabilidade de ter a doença.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

# Criando as IAs (modelos)
modelo_arvore = DecisionTreeClassifier(random_state=42)
modelo_regressao = LogisticRegression(max_iter=500, random_state=42)

print("Modelos importados e prontos para uso!")
```

### Etapa 3: Treinamento utilizando K-Fold
O K-Fold é um teste mais rigoroso. Em vez de testar a IA uma única vez, nós dividimos os dados em 5 partes (K=5). A IA treina e testa 5 vezes, revezando as partes. Isso garante que o resultado seja confiável.

```python
from sklearn.model_selection import cross_validate

# Separamos o que queremos prever (Resultado) do resto (Características)
X = df.drop('Resultado', axis=1) # Todas as colunas, menos o Resultado
y = df['Resultado']              # Apenas a coluna Resultado (0 ou 1)

# Definimos as métricas que queremos avaliar
metricas = ['accuracy', 'precision', 'recall', 'f1']

# Treinando e avaliando a Árvore com K-Fold (5 rodadas)
resultados_arvore = cross_validate(modelo_arvore, X, y, cv=5, scoring=metricas)

# Treinando e avaliando a Regressão com K-Fold (5 rodadas)
resultados_regressao = cross_validate(modelo_regressao, X, y, cv=5, scoring=metricas)

print("Treinamento K-Fold concluído!")
```

### Etapa 4: Comparação dos modelos e Métricas
Explicação das Métricas:

Acurácia (Accuracy): Acertos totais. Quantos diagnósticos (positivos e negativos) o modelo acertou no geral?

Precisão (Precision): Qualidade do alerta. De todas as vezes que o modelo disse "tem diabetes", quantas ele realmente estava certo?

Recall (Sensibilidade/Score): Capacidade de encontrar. De todas as pessoas que realmente têm diabetes, quantas o modelo conseguiu identificar?

F1-Score: Equilíbrio. É a média equilibrada entre a Precisão e o Recall.

```python
# Função para calcular e exibir a média dos resultados do K-Fold
def mostrar_resultados(nome_modelo, resultados):
    print(f"--- Resultados: {nome_modelo} ---")
    print(f"Acurácia: {resultados['test_accuracy'].mean():.2f}")
    print(f"Precisão: {resultados['test_precision'].mean():.2f}")
    print(f"Recall:   {resultados['test_recall'].mean():.2f}")
    print(f"F1-Score: {resultados['test_f1'].mean():.2f}\n")

# Exibindo a comparação
mostrar_resultados("Árvore de Decisão", resultados_arvore)
mostrar_resultados("Regressão Logística", resultados_regressao)
```
