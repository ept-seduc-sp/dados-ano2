```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

# y_verdadeiro representa a realidade (0 = Email Normal, 1 = Spam)
# y_previsto representa o que o modelo deduziu
y_verdadeiro = [0, 1, 1, 0, 1, 0]
y_previsto   = [0, 1, 0, 0, 1, 1]

print("Acurácia:", accuracy_score(y_verdadeiro, y_previsto))
print("Precisão:", precision_score(y_verdadeiro, y_previsto))
print("Revocação:", recall_score(y_verdadeiro, y_previsto))
```
