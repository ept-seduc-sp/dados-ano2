## Cenário 1 — Treinando o Modelo A: Regressão Logística

````python
# Passos 1 e 2: Importação e Preparação de Dados
import pandas as pd

df = pd.read_csv('Dataset_Deteccao_Fraude_Pagamentos_Online_pt.csv')
df = df.drop(columns=['nomeOrigem', 'nomeDestino'])
df = pd.get_dummies(df, columns=['tipo'])

X = df.drop('fraude', axis=1)
y = df['fraude']

# Passo 3: Separação de Dados
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Passo 4: Treinamento do Modelo A
from sklearn.linear_model import LogisticRegression
# Nota: max_iter=1000 adicionado para evitar avisos de limite de iteração
modelo_a = LogisticRegression(max_iter=1000)
modelo_a.fit(X_train, y_train)
previsoes_A = modelo_a.predict(X_test)

# Passo 5: Avaliação de Métricas
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Resultados do Modelo A (Regressão Logística):")
print("Acurácia:", accuracy_score(y_test, previsoes_A))
print("Precisão:", precision_score(y_test, previsoes_A))
print("Revocação (Recall):", recall_score(y_test, previsoes_A))
print("F1-Score:", f1_score(y_test, previsoes_A))
````

## Cenário 2 — Treinando o Modelo B e Tomando a Decisão (Foco em Recall)

````python
# Passo 1: Importação
from sklearn.ensemble import RandomForestClassifier

# Passo 2: Treinamento do Modelo B
modelo_b = RandomForestClassifier(random_state=42)
modelo_b.fit(X_train, y_train)
previsoes_B = modelo_b.predict(X_test)

# Passo 3: Avaliação de Métricas
print("Resultados do Modelo B (Random Forest):")
print("Acurácia:", accuracy_score(y_test, previsoes_B))
print("Precisão:", precision_score(y_test, previsoes_B))
print("Revocação (Recall):", recall_score(y_test, previsoes_B))
print("F1-Score:", f1_score(y_test, previsoes_B))
````
