### Etapa 1: Preparação do Ambiente e Dados
Antes de iniciar, importamos as bibliotecas necessárias. Vamos simular os dados de 12 meses de uma operação de e-commerce (como o Mercado Livre), onde X é o volume de vendas e y é o tempo médio de entrega. Meses de pico, como a Black Friday, têm maior volume e prazos diferentes.

```python
import numpy as np
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# Dados: 12 meses do ano
meses = np.array(['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez'])

# X = Volume de pacotes (em milhares)
X = np.array([10, 12, 11, 15, 14, 16, 15, 18, 20, 22, 50, 45]).reshape(-1, 1)

# y = Tempo médio de entrega (em dias)
y = np.array([2, 2, 2, 3, 3, 3, 3, 4, 4, 5, 10, 8])

# Iniciando o modelo de Machine Learning
modelo = LinearRegression()
print("Dados e modelo carregados com sucesso!")
```

### Etapa 2: Configurando o K-Fold

Vamos definir nosso rodízio para garantir que o modelo não apenas "decore" meses tranquilos, mas saiba generalizar para picos de logística. Criaremos 4 fatias (K=4).

```python
# Criando 4 "fatias" (K=4)
kf = KFold(n_splits=4, shuffle=False)
print("Configuração do K-Fold concluída.")
```

### Etapa 3: Executando o Rodízio e Treinando o Modelo (Loop)

Agora faremos o modelo rodar pelas fatias. Em cada rodada, ele treina com uma parte do ano e testa na parte separada. Calculamos o erro médio em dias de entrega para ver o desempenho real.

```python
# Mostrando o rodízio e o treinamento na prática
rodada = 1
for treino_idx, teste_idx in kf.split(meses):
    print(f"--- Rodada {rodada} ---")
    print(f"Treino: {meses[treino_idx]} | Teste: {meses[teste_idx]}")
    
    # Separando os dados
    X_treino, X_teste = X[treino_idx], X[teste_idx]
    y_treino, y_teste = y[treino_idx], y[teste_idx]
    
    # Treinando o modelo
    modelo.fit(X_treino, y_treino)
    
    # Fazendo a previsão
    previsoes = modelo.predict(X_teste)
    
    # Avaliando o erro (em dias)
    erro = mean_absolute_error(y_teste, previsoes)
    print(f"Erro médio na previsão de entrega: {erro:.2f} dias\n")
    
    rodada += 1
```
