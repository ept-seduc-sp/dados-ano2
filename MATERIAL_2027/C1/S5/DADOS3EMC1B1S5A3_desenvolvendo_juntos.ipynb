### Etapa 1: Preparação do Ambiente e Leitura dos Dados
Antes de prever, precisamos carregar os dados reais da distribuidora (arquivo `cocacola_enterprise_dataset_pt_br.csv`) e separar nossas variáveis. Aqui, o Gasto em Marketing é o nosso "motor" de vendas.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Carregando o dataset real
df = pd.read_csv("cocacola_enterprise_dataset_pt_br.csv")

# Separando: Gasto em Marketing (X) e Unidades Vendidas (y)
X = df[['Gasto_Marketing']] 
y = df['Unidades_Vendidas']

print("Dados carregados com sucesso! Prontos para análise.")
```

### Etapa 2: Visualizando a Relação das Variáveis

Vamos gerar um gráfico de dispersão para ver como o investimento em marketing afetou as vendas no passado.

```python
# Gráfico de pontos
plt.scatter(X, y, color='green')
plt.title('Unidades Vendidas vs Gasto em Marketing')
plt.xlabel('Gasto em Marketing (R$)')
plt.ylabel('Unidades Vendidas')
plt.show()
```

### Etapa 3: Treinando o Modelo e Traçando a Relação Linear
Agora, aplicamos a regressão linear para achar a reta que melhor explica esse comportamento.

```python
# Criando e treinando a IA (Regressão Linear)
modelo = LinearRegression()
modelo.fit(X, y)

# Gerando a reta de previsão no gráfico
plt.scatter(X, y, color='green')
plt.plot(X, modelo.predict(X), color='blue') # A reta azul de tendência!
plt.title('Modelo de Regressão Linear - Impacto do Marketing')
plt.xlabel('Gasto em Marketing (R$)')
plt.ylabel('Previsão de Unidades Vendidas')
plt.show()

# Fazendo a previsão para um investimento de R$ 3.000,00!
previsao = modelo.predict([[3000]])
print(f"Para um investimento de R$ 3000 em marketing, o modelo sugere a venda de: {int(previsao[0])} unidades.")
```
