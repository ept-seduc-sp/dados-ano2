<h2 align="center">Previsão de Receita com Regressão Linear</h2>

## Base Teórica

**Exemplo de código:**

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Criando dados de exemplo
dados = {'Propaganda_R$': [100, 200, 300, 400], 'Vendas_Unidades': [10, 15, 20, 25]}
df = pd.DataFrame(dados)

# Separando variáveis
X = df[['Propaganda_R$']] # Variável preditora (sempre em 2D)
y = df['Vendas_Unidades'] # Variável alvo

# Treinando o modelo
modelo = LinearRegression()
modelo.fit(X, y)

# Prevendo vendas para um investimento de R$ 250
previsao = modelo.predict([[250]])
print(f"Previsão de vendas: {previsao[0]:.2f} unidades")
```

## Cenário 1 — Previsão de Receita com Base no Clima

````python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Passo 2
df = pd.read_csv('Ice_Cream.csv')
print(df.head())

# Passo 3
X = df[['Temperatura']]
y = df['Receita']

# Passo 4
modelo = LinearRegression()
modelo.fit(X, y)

# Passo 5
previsoes = modelo.predict(X)
````

## Cenário 2 — Interpretando o Modelo e Seus Limites

````python
# Passo 1
df_comparacao = pd.DataFrame({'Real': y, 'Previsto': previsoes})
print(df_comparacao.head())

# Passo 2
previsao_frio = modelo.predict([[1]])
previsao_calor = modelo.predict([[45]])
print(f"Previsão 1°C: {previsao_frio[0]:.2f}")
print(f"Previsão 45°C: {previsao_calor[0]:.2f}")
````

* Não, os valores são próximos, mas não exatos. Isso significa que o modelo gera estimativas baseadas na tendência geral, e não certezas absolutas, existindo sempre uma margem de erro.

* Extrapolar para temperaturas muito distantes dos dados originais (como 45°C) é arriscado. O gerente não deve confiar cegamente no modelo para qualquer temperatura, pois a regressão linear segue uma linha reta infinita que pode gerar valores irreais que fogem da lógica física do negócio e do comportamento humano em climas extremos.
