### Etapa 1: Importação de Bibliotecas e Dados
Antes de simplificar, precisamos das ferramentas do scikit-learn e de um dataset clínico. Usaremos o dataset Breast Cancer, que contém exames de 569 pacientes. Ele possui 30 variáveis numéricas (como raio, textura e área do tumor) e o diagnóstico classificando o tumor como benigno ou maligno.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import load_breast_cancer

# Carregando dados
dados = load_breast_cancer()
X = dados.data # 30 variáveis originais
y = dados.target # Diagnóstico

print(f"Formato original dos dados: {X.shape}")

# Convertendo para DataFrame para visualizar com o head()
df = pd.DataFrame(X, columns=dados.feature_names)
print("\nVisualizando as 5 primeiras linhas:")
print(df.head())
```

### Etapa 2: Aplicando a Redução de Dimensionalidade (PCA)

Vamos reduzir as 30 variáveis para apenas 2 componentes principais, descartando o "ruído" e mantendo a essência.

```python
# Criando o redutor para 2 dimensões
pca = PCA(n_components=2)

# Transformando os dados
X_reduzido = pca.fit_transform(X)
print(f"Novo formato dos dados: {X_reduzido.shape}") # (569 pacientes, 2 atributos)
```

### Etapa 3: Visualizando o Padrão

Agora que temos apenas 2 variáveis, podemos criar um gráfico 2D simples para ver se os dados continuam fazendo sentido médico.

```python
plt.scatter(X_reduzido[:, 0], X_reduzido[:, 1], c=y, cmap='coolwarm', alpha=0.7)
plt.title('Visualização de Dados Clínicos após Redução')
plt.xlabel('Componente Essencial 1')
plt.ylabel('Componente Essencial 2')
plt.colorbar(label='Diagnóstico')
plt.show()
```
