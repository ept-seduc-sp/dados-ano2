### Etapa 1: Importação, Dados e Padronização
Antes de fazer a redução, importamos as ferramentas e o dataset clínico (569 pacientes, 30 variáveis). Passo crucial: padronizamos os dados com StandardScaler. Como o PCA é sensível a escalas, variáveis grandes ofuscariam variáveis menores. A qualidade da informação depende dessa padronização prévia.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer

# Carregando dados
dados = load_breast_cancer()
X = dados.data      # 30 variáveis originais
y = dados.target    # Diagnóstico

print(f"Formato original dos dados: {X.shape}")

# Padronizando os dados 
scaler = StandardScaler()
X_escalado = scaler.fit_transform(X)

# Convertendo para DataFrame para visualização
df = pd.DataFrame(X_escalado, columns=dados.feature_names)
print("\nVisualizando as 5 primeiras linhas:")
print(df.head())
```

### Etapa 2: Aplicando a Redução de Dimensionalidade (PCA) com Meta de Variância

Em vez de apenas definir "2 componentes", delegamos a decisão ao PCA estabelecendo uma Meta de Variância.

Por que 95% (0.95)? Como estamos lidando com dados clínicos (câncer de mama), a qualidade e retenção da informação são críticas. O valor de 95% garante que quase toda a variabilidade (comportamento) dos dados originais seja mantida, eliminando apenas 5% referentes a redundâncias ou ruídos. Se os dados tivessem baixa qualidade ou extremo ruído (como sensores defeituosos), usaríamos uma meta menor (ex: 80% ou 85%).

```python
# Criando o redutor com Meta de Variância de 95%
pca = PCA(n_components=0.95)

# Transformando os dados
X_reduzido = pca.fit_transform(X_escalado)

print(f"O PCA determinou que precisamos de {pca.n_components_} componentes para reter 95% da informação.")
print(f"Novo formato dos dados: {X_reduzido.shape}")
```

### Etapa 3: Visualizando o peso de todos os Componentes
Também podemos usar um gráfico de barras com o método explained_variance_ratio_ para entender a importância de cada componente criado para atingir a meta de 95%.

```python
import numpy as np

# Calculando a porcentagem de variância de cada componente
variancia = pca.explained_variance_ratio_ * 100

# Criando o gráfico
plt.bar(range(1, pca.n_components_ + 1), variancia, color='teal')
plt.title('Importância de cada Componente Principal')
plt.xlabel('Número do Componente')
plt.ylabel('Informação Retida (%)')
plt.xticks(range(1, pca.n_components_ + 1))
plt.show()
```
