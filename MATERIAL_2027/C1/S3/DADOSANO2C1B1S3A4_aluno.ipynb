## Base Teórica

**Exemplo de código:**

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import seaborn as sns

# 1. Dados de exemplo (diferentes do cenário prático)
dados = pd.DataFrame({
    'Visitantes': [150, 160, 140, 500, 520, 490],
    'Vendas_Reais': [1500, 1650, 1400, 8000, 8500, 7900]
})

# 2. Padronizando os dados
scaler = StandardScaler()
dados_padronizados = scaler.fit_transform(dados)

# 3. Criando e aplicando o K-Means para 2 grupos
kmeans = KMeans(n_clusters=2, random_state=42)
dados['Cluster'] = kmeans.fit_predict(dados_padronizados)

# 4. Visualizando o resultado
sns.scatterplot(data=dados, x='Visitantes', y='Vendas_Reais', hue='Cluster', palette='viridis')
```
