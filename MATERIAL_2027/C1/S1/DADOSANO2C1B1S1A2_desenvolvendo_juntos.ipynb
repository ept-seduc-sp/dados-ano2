### Etapa 1: Importando a biblioteca e criando os dados brutos

Para começarmos, vamos importar o pandas, a biblioteca padrão para manipulação de dados em Python. Vamos simular os dados de clientes de uma loja.

```python
import pandas as pd

# Nossos dados brutos do sistema da loja
dados_loja = {
    'Idade': [25, 42, 31, 50, 19],
    'Gasto_Mensal': [150.00, 450.50, 200.00, 600.00, 80.00],
    'Comprou_Promocao': ['Sim', 'Não', 'Sim', 'Não', 'Sim']
}
print("Dados brutos definidos!")
```

### Etapa 2: Separando as Características (Features)

Em Ciência de Dados, as características costumam ser chamadas de "X". Elas são as variáveis que usamos para tentar prever algo. Aqui, a idade e o gasto mensal ditam o comportamento do cliente.

```python
df = pd.DataFrame(dados_loja)

# Separando apenas as características
caracteristicas = df[['Idade', 'Gasto_Mensal']]
print("Características (X):")
print(caracteristicas)
```

### Etapa 3: Identificando o Rótulo (Label) e Dados Não Rotulados

O rótulo, geralmente chamado de "y", é a resposta que queremos prever. Se a coluna "Comprou_Promocao" existir, temos um dado rotulado (Supervisionado). Se a removermos, temos um dado não rotulado (Não Supervisionado).

```python
# Isolando o Rótulo (y)
rotulo = df['Comprou_Promocao']
print("\nRótulo (y):")
print(rotulo)

# Simulando um conjunto Não Rotulado (pronto para agrupamento)
dados_nao_rotulados = df.drop(columns=['Comprou_Promocao'])
print("\nDados Não Rotulados (Para Aprendizado Não Supervisionado):")
print(dados_nao_rotulados)
```
