## Gabarito — Classificador de Alerta de Baixa Umidade

````python
import pandas as pd
import numpy as np

# Passo 1 e 2: Importar pandas e ler o CSV
df_clima = pd.read_csv('INMET_CO_DF_A001_BRASILIA_01-01-2026_A_31-05-2026.CSV', sep=';', decimal=',')

# Passo 3: Visualizar as colunas
display(df_clima.head())

# Passo 4: Criar coluna de alerta
# Pode ser feito usando list comprehension ou np.where
df_clima['Alerta_Baixa_Umidade'] = np.where(df_clima['Umidade Relativa Mínima (%)'] <= 30, 'ALERTA', 'NORMAL')

# Passo 5: Contagem
contagem_alertas = df_clima['Alerta_Baixa_Umidade'].value_counts()
print(contagem_alertas)
````

## Gabarito — Tratamento de Nulos

````python
# Passo 1 e 2: Preencher valores nulos
df_clima = df_clima.dropna(subset=['Vento, Rajada Máxima (m/s)'])

# Passo 3 e 4: Criar regra de classificação do vento
df_clima['Risco_Vento'] = ['Risco Alto' if vento > 5.0 else 'Risco Baixo' for vento in df_clima['Vento, Rajada Máxima (m/s)']]

# Passo 5: Filtrar e mostrar os padrões de risco alto
risco_alto_df = df_clima[df_clima['Risco_Vento'] == 'Risco Alto']
display(risco_alto_df[['Data', 'Hora UTC', 'Vento, Rajada Máxima (m/s)', 'Risco_Vento']].head(10))
````
