### Etapa 1: Preparação e Separação de Dados

Sempre precisamos dividir nossos dados em duas partes: Treino (para a IA aprender) e Teste (para vermos se ela aprendeu de verdade com dados que ela nunca viu).

```python
import pandas as pd
from sklearn.model_selection import train_test_split


df_bank = pd.read_csv('Bank_Marketing_Dataset_PT.csv')
df_sleep = pd.read_csv('student_sleep_mental_health_2026_PT.csv')

print("Bases carregadas com sucesso!")
```

### Etapa 2: Caso Nubank (Classificação e Avaliação)

Vamos prever se o cliente assinará o depósito. Usaremos 3 métricas vitais:

Acurácia: Do total, quantos a IA acertou?

Precisão: Quando a IA disse que o cliente ia assinar, quantas vezes ela realmente acertou?

Recall (Revocação): De todos os clientes que realmente assinaram, quantos a IA conseguiu encontrar?

Python

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score

# 1. Separando características (X) e o alvo (y)
X_bank = df_bank[['idade', 'saldo_anual', 'duracao_segundos']] 
y_bank = df_bank['assinou_deposito']

# 2. Dividindo os dados: 70% para treino e 30% para teste
X_treino_b, X_teste_b, y_treino_b, y_teste_b = train_test_split(X_bank, y_bank, test_size=0.3, random_state=42)

# 3. Criando e treinando o modelo de Classificação
modelo_classificacao = DecisionTreeClassifier(random_state=42)
modelo_classificacao.fit(X_treino_b, y_treino_b)

# 4. Avaliando o modelo com os dados de teste
previsoes_teste_b = modelo_classificacao.predict(X_teste_b)

print("--- Avaliação do Modelo Nubank ---")
print(f"Acurácia: {accuracy_score(y_teste_b, previsoes_teste_b):.2f}")
# Usamos pos_label='sim' pois queremos medir o acerto da resposta positiva
print(f"Precisão: {precision_score(y_teste_b, previsoes_teste_b, pos_label='sim', zero_division=0):.2f}")
print(f"Recall: {recall_score(y_teste_b, previsoes_teste_b, pos_label='sim', zero_division=0):.2f}")
```

### Etapa 4: Caso Alice Saúde (Regressão e Avaliação)

Aqui prevemos um número (horas de sono). Usaremos 3 métricas:

MAE: Erro médio absoluto (quantas horas erramos em média).

MSE: Erro quadrático médio (penaliza erros grandes).

R²: O quanto nosso modelo explica os dados (mais próximo de 1, melhor).

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Separando características (X) e o alvo (y)
X_sleep = df_sleep[['horas_estudo_por_dia', 'horas_tempo_tela']] 
y_sleep = df_sleep['media_horas_sono']

# 2. Dividindo os dados (70% treino, 30% teste)
X_treino_s, X_teste_s, y_treino_s, y_teste_s = train_test_split(X_sleep, y_sleep, test_size=0.3, random_state=42)

# 3. Criando e treinando o modelo de Regressão
modelo_regressao = LinearRegression()
modelo_regressao.fit(X_treino_s, y_treino_s)

# 4. Avaliando o modelo
previsoes_teste_s = modelo_regressao.predict(X_teste_s)

print("--- Avaliação do Modelo Alice Saúde ---")
print(f"Erro Médio Absoluto (MAE): {mean_absolute_error(y_teste_s, previsoes_teste_s):.2f} horas")
print(f"Erro Quadrático Médio (MSE): {mean_squared_error(y_teste_s, previsoes_teste_s):.2f}")
print(f"Score R²: {r2_score(y_teste_s, previsoes_teste_s):.2f}")
```

### Etapa 5: Colocando os Modelos em Ação (Previsões Finais)

Fazendo previsões para novos perfis criados por nós.

```python
# ==========================================
# 1. Novo Cliente - Nubank
# ==========================================
dados_novo_cliente = pd.DataFrame({
    'idade': [35], 
    'saldo_anual': [1500], 
    'duracao_segundos': [300]
})

previsao_deposito = modelo_classificacao.predict(dados_novo_cliente)
print(f"\nPrevisão Nubank: O cliente fará o depósito? '{previsao_deposito[0]}'")

# ==========================================
# 2. Novo Estudante - Alice Saúde
# ==========================================
dados_novo_estudante = pd.DataFrame({
    'horas_estudo_por_dia': [4], 
    'horas_tempo_tela': [6]
})

previsao_horas_sono = modelo_regressao.predict(dados_novo_estudante)
print(f"Previsão Alice Saúde: O estudante dorme em média: {previsao_horas_sono[0]:.1f} horas")
```
