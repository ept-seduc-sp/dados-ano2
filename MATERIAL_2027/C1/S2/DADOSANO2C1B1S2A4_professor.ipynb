## Base Teórica

**Exemplo de código:**

Exemplo de Regressão: Prevendo produtividade agrícola (kg) com base em horas de sol.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dados de exemplo
dados_sol = pd.DataFrame({
    'horas_sol': [4, 5, 6, 7, 8], 
    'produtividade': [100, 120, 150, 170, 190]
})
X = dados_sol[['horas_sol']] # Variável preditora
y = dados_sol['produtividade'] # Variável alvo contínua

# Treinamento do modelo
modelo_regressao = LinearRegression()
modelo_regressao.fit(X, y)

# Previsão para um dia com 6.5 horas de sol
previsao = modelo_regressao.predict([[6.5]])
print(f"Previsão de produtividade: {previsao[0]:.2f} kg")
```

Exemplo de Classificação: Categorizando se uma planta precisa de irrigação (1 = Sim, 0 = Não) baseado na umidade do solo.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Dados de exemplo
dados_umidade = pd.DataFrame({
    'umidade_solo': [20, 30, 60, 80], 
    'precisa_agua': [1, 1, 0, 0]
})
X_class = dados_umidade[['umidade_solo']] # Variável preditora
y_class = dados_umidade['precisa_agua'] # Variável alvo categórica

# Treinamento do modelo
modelo_classificacao = DecisionTreeClassifier()
modelo_classificacao.fit(X_class, y_class)

# Previsão para um solo com 40% de umidade
previsao_agua = modelo_classificacao.predict([[40]])
print(f"Precisa de água? (1=Sim, 0=Não): {previsao_agua[0]}")
```

## Gabarito — Prevendo Valores de Propriedades (Regressão)

````python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Passos 1 e 2
df_imoveis = pd.read_csv('dubai_residential_data_2026_pt.csv')
print(df_imoveis.head())

# Passos 3 e 4
# Usando apenas uma coluna numérica para simplificar a didática e evitar transformações
X = df_imoveis[['AREA_UTIL_M2']] 
y = df_imoveis['VALOR_TRANSACAO']

# Remover possíveis linhas vazias para evitar erros no fit
df_treino = pd.concat([X, y], axis=1).dropna()
X_treino = df_treino[['AREA_UTIL_M2']]
y_treino = df_treino['VALOR_TRANSACAO']

# Passos 5 e 6
modelo_reg = LinearRegression()
modelo_reg.fit(X_treino, y_treino)

# Passo 7
novo_imovel = pd.DataFrame({'AREA_UTIL_M2': [120]})
previsao_preco = modelo_reg.predict(novo_imovel)
print(f"Valor estimado: {previsao_preco[0]:.2f}")
````

## Gabarito — Categorizando Imóveis de Alto Padrão (Classificação)

````python
from sklearn.tree import DecisionTreeClassifier

# Passo 1
df_imoveis['categoria_luxo'] = (df_imoveis['VALOR_TRANSACAO'] > 2500000).astype(int)

# Passos 2 e 3
X_class = df_imoveis[['AREA_UTIL_M2']]
y_class = df_imoveis['categoria_luxo']

# Remover valores nulos, caso existam
df_treino_class = pd.concat([X_class, y_class], axis=1).dropna()
X_class_treino = df_treino_class[['AREA_UTIL_M2']]
y_class_treino = df_treino_class['categoria_luxo']

# Passos 4 e 5
modelo_class = DecisionTreeClassifier()
modelo_class.fit(X_class_treino, y_class_treino)

# Passo 6
imovel_marketing = pd.DataFrame({'AREA_UTIL_M2': [200]})
previsao_luxo = modelo_class.predict(imovel_marketing)
print(f"Categoria VIP (1=Sim, 0=Não): {previsao_luxo[0]}")
````
