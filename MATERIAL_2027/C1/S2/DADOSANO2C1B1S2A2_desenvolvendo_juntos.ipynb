### Etapa 1: Importação de Bibliotecas e Estruturação de Dados

Antes de treinar o modelo, precisamos importar o classificador e organizar nossos arrays de dados.

```python
from sklearn.tree import DecisionTreeClassifier

# Definindo nossos dados de treinamento (Renda, Atrasos)
x_treino = [
    [5000, 0], 
    [1500, 4], 
    [8000, 0], 
    [2500, 2]
]

# Definindo os rótulos (Labels) para cada cliente
y_treino = ['Aprovado', 'Reprovado', 'Aprovado', 'Reprovado']

print("Dados de treinamento organizados e prontos!")
```

### Etapa 2: Instanciando e Treinando o Modelo

Vamos criar a "inteligência" (o modelo) e ensiná-la a relacionar as características às classes.

```python
# Instanciando o modelo de Árvore de Decisão
classificador = DecisionTreeClassifier()

# O comando fit realiza o treinamento supervisionado
classificador.fit(x_treino, y_treino)

print("Modelo treinado com sucesso!")
```

### Etapa 3: Realizando uma Classificação (Previsão Categórica)

Agora colocamos o modelo à prova com dados de um cliente inédito.

```python
# Dados do novo cliente: Renda de R$ 3200, 1 atraso
novo_cliente = [[3200, 1]]

# O comando predict determina a classe
decisao = classificador.predict(novo_cliente)

print(f"Decisão algorítmica para o novo cliente: {decisao[0]}")
```

### Etapa 4: Avaliando o Desempenho do Modelo

Para sabermos se a nossa IA realmente aprendeu, precisamos testá-la com um novo conjunto de clientes (dados de teste) e comparar as respostas do modelo com a realidade usando métricas de avaliação.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

# 1. Criando os dados de teste (novos clientes) e o "gabarito" (resultado real)
x_teste = [
    [7000, 0], # Cliente A: Renda alta, sem atrasos
    [2000, 3], # Cliente B: Renda baixa, com atrasos
    [3000, 1]  # Cliente C: Renda média, 1 atraso
]
y_real = ['Aprovado', 'Reprovado', 'Aprovado'] # O que realmente aconteceu na vida real

# 2. Pedindo para o modelo fazer as previsões para esses novos clientes
y_previsao = classificador.predict(x_teste)

# 3. Calculando as métricas 
# Obs: Usamos pos_label='Aprovado' para dizer ao Python qual é a classe principal que estamos avaliando
acuracia = accuracy_score(y_real, y_previsao)
precisao = precision_score(y_real, y_previsao, pos_label='Aprovado')
revocacao = recall_score(y_real, y_previsao, pos_label='Aprovado')

print("--- RESULTADOS DA AVALIAÇÃO ---")
print(f"Acurácia: {acuracia * 100}% (O quanto o modelo acertou no total)")
print(f"Precisão: {precisao * 100}% (Quando o modelo disse 'Aprovado', ele estava certo em {precisao * 100}%)")
print(f"Revocação: {revocacao * 100}% (De todos os 'Aprovados' reais, o modelo encontrou {revocacao * 100}%)")
```
