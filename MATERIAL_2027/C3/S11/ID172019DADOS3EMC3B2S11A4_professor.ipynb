```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib import font_manager
from IPython.display import display

arquivo_csv = Path("DADOS3EMC3B2S11A4_custos_frete.csv")
if not arquivo_csv.is_file():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {arquivo_csv}. Salve o CSV e o notebook na mesma pasta."
    )

dados = pd.read_csv(arquivo_csv)
referencia_operacional = 24.0

# Padrão visual: usa Poppins quando a fonte estiver disponível no ambiente.
if any(f.name == "Poppins" for f in font_manager.fontManager.ttflist):
    plt.rcParams["font.family"] = "Poppins"
plt.rcParams["font.size"] = 26

print("Dados da atividade:")
display(dados)
print(f"Referência operacional: R$ {referencia_operacional:,.2f} por pedido".replace(",", "X").replace(".", ",").replace("X", "."))
```

## Etapa 4 — Produzam a versão aprimorada

A célula de apoio abaixo contém a função que desenha o gráfico. **Não é necessário alterá-la.**

Na célula seguinte, editem somente as configurações que correspondem às decisões registradas pela dupla.


```python
def plotar_visualizacao(
    tabela,
    ordem_centros,
    orientacao,
    limite_inferior,
    limite_superior,
    titulo,
    mostrar_referencia,
    mostrar_valores,
    mostrar_legenda,
    centros_destacados,
    cor_base,
    cor_destaque,
):
    tabela_plot = tabela.set_index("centro_distribuicao").loc[ordem_centros].reset_index()
    valores = tabela_plot["custo_medio_frete_por_pedido"].tolist()
    cores = [cor_destaque if centro in centros_destacados else cor_base for centro in ordem_centros]

    fig, ax = plt.subplots(figsize=(18, 10))

    if orientacao == "horizontal":
        barras = ax.barh(ordem_centros, valores, color=cores)
        ax.set_xlabel("Custo médio de frete por pedido (R$)", fontsize=26)
        ax.set_xlim(limite_inferior, limite_superior)
        if mostrar_referencia:
            ax.axvline(referencia_operacional, color="#312F67", linewidth=2, linestyle="--")
        if mostrar_valores:
            for barra, valor in zip(barras, valores):
                ax.text(valor + 0.15, barra.get_y() + barra.get_height()/2,
                        f"R$ {valor:.2f}".replace(".", ","), va="center", fontsize=26)
    else:
        barras = ax.bar(ordem_centros, valores, color=cores)
        ax.set_ylabel("Custo médio de frete por pedido (R$)", fontsize=26)
        ax.set_ylim(limite_inferior, limite_superior)
        ax.tick_params(axis="x", rotation=28, labelsize=26)
        if mostrar_referencia:
            ax.axhline(referencia_operacional, color="#312F67", linewidth=2, linestyle="--")
        if mostrar_valores:
            for barra, valor in zip(barras, valores):
                ax.text(barra.get_x() + barra.get_width()/2, valor + 0.15,
                        f"R$ {valor:.2f}".replace(".", ","), ha="center", fontsize=26)

    ax.set_title(titulo, fontsize=30, weight="bold")
    ax.tick_params(axis="both", labelsize=26)
    ax.grid(axis="x" if orientacao == "horizontal" else "y", alpha=0.20)

    if mostrar_legenda:
        ax.legend(
            handles=[
                Patch(facecolor=cor_base, label="Demais centros"),
                Patch(facecolor=cor_destaque, label="Centros destacados")
            ],
            loc="best",
            fontsize=26
        )

    plt.tight_layout()
    plt.show()
```

### Configuração da visualização aprimorada

A configuração abaixo começa com escolhas semelhantes às da visualização inicial para que a célula execute sem edição. **Altere somente o que a dupla decidiu modificar.**

Vocês podem mudar ordem, orientação, escala, título, referência, rótulos, legenda e destaques. Não existe uma combinação única obrigatória.


```python
# EDITE AQUI de acordo com o diagnóstico e o planejamento da dupla.
ordem_centros = dados.sort_values("centro_distribuicao")["centro_distribuicao"].tolist()
orientacao = "vertical"            # "vertical" ou "horizontal"
limite_inferior = 20
limite_superior = 30

titulo = "Custo médio de frete por CD"
mostrar_referencia = False
mostrar_valores = False
mostrar_legenda = True

centros_destacados = []
cor_base = "#AAA6D3"
cor_destaque = "#FF161F"

plotar_visualizacao(
    dados,
    ordem_centros,
    orientacao,
    limite_inferior,
    limite_superior,
    titulo,
    mostrar_referencia,
    mostrar_valores,
    mostrar_legenda,
    centros_destacados,
    cor_base,
    cor_destaque,
)
```

```python
# Conferência computacional do professor — uma solução possível, não obrigatória
ordem_referencia = dados.sort_values("custo_medio_frete_por_pedido", ascending=True)["centro_distribuicao"].tolist()
centros_acima = dados.loc[
    dados["custo_medio_frete_por_pedido"] > referencia_operacional,
    "centro_distribuicao"
].tolist()

plotar_visualizacao(
    dados,
    ordem_centros=ordem_referencia,
    orientacao="horizontal",
    limite_inferior=0,
    limite_superior=32,
    titulo="Três centros ultrapassam a referência de R$ 24 por pedido",
    mostrar_referencia=True,
    mostrar_valores=True,
    mostrar_legenda=False,
    centros_destacados=centros_acima,
    cor_base="#AAA6D3",
    cor_destaque="#FF161F",
)
```
