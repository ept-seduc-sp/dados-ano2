Executem as verificações preparadas. Concentrem a interpretação nas variáveis definidas nesta atividade.

```python
# Ausências nas variáveis selecionadas
resumo_ausencias = pd.DataFrame({
    "valores_ausentes": dados[variaveis_selecionadas].isna().sum(),
    "percentual_ausente": (100 * dados[variaveis_selecionadas].isna().mean()).round(2)
})
display(resumo_ausencias)
for coluna in variaveis_selecionadas:
    if dados[coluna].isna().any():
        print(f"Exemplos de registros com ausência em {coluna} (até 10):")
        campos = list(dict.fromkeys(["id_registro", "data", "loja", coluna]))
        display(dados.loc[dados[coluna].isna(), campos].head(10))

# Duplicação integral: conferir todas as colunas, preservando os identificadores.
copias_adicionais = int(dados.duplicated().sum())
duplicados_integrais = dados.loc[dados.duplicated(keep=False)].sort_values("id_registro")
print("Cópias integrais adicionais detectadas:", copias_adicionais)
print("Todas as ocorrências envolvidas na duplicação integral:")
display(duplicados_integrais)
```

## Etapa 3 — Examinem tipos, formatos e consistência — 9 minutos

Observem como as colunas definidas para esta atividade foram lidas e se a representação é compatível com seu significado e sua unidade. O tipo computacional, isoladamente, não comprova erro.

A célula mostra as representações das duas variáveis de despesas e examina valores não ausentes fora de 0 a 100 em `desconto_campanha_pct` e `disponibilidade_produtos_pct`. Um resultado sem ocorrência detectada também deve ser registrado.

```python
# Tipos e ausências nas colunas selecionadas
resumo_tipos = pd.DataFrame({
    "tipo_lido": dados[variaveis_selecionadas].dtypes.astype(str),
    "valores_ausentes": dados[variaveis_selecionadas].isna().sum()
})
display(resumo_tipos)

colunas_despesas = ["despesas_com_pessoal", "outras_despesas_operacionais"]
for coluna in colunas_despesas:
    # Representação temporária para inspeção; a coluna original é preservada.
    valores = dados[coluna].dropna().astype(str)
    resumo_formatos = pd.DataFrame({
        "verificacao": ["valores não ausentes", "valores contendo R$",
                        "valores contendo vírgula", "valores contendo ponto"],
        "quantidade": [len(valores), int(valores.str.contains(r"R\$", regex=True).sum()),
                       int(valores.str.contains(",", regex=False).sum()),
                       int(valores.str.contains(".", regex=False).sum())]
    })
    print(f"Representações observadas em {coluna}:")
    display(resumo_formatos)
    mascara_diferente = valores.str.contains(r"R\$|,", regex=True)
    print("Exemplos para conferência (até 10):")
    display(dados.loc[valores[mascara_diferente].index, ["id_registro", coluna]].head(10))

colunas_percentuais = ["desconto_campanha_pct", "disponibilidade_produtos_pct"]
checagens = []
for coluna in colunas_percentuais:
    fora_da_faixa = dados[coluna].notna() & ~dados[coluna].between(0, 100)
    checagens.append({"verificacao": f"{coluna}: valores não ausentes fora de 0 a 100",
                     "ocorrencias": int(fora_da_faixa.sum())})
    if fora_da_faixa.any():
        display(dados.loc[fora_da_faixa, ["id_registro", coluna]].head(10))
print("Verificações de faixa; ausências são registradas separadamente na etapa 2:")
display(pd.DataFrame(checagens))
```

## Etapa 4 — Examinem possíveis valores atípicos — 10 minutos

Executem a célula para observar o resumo das cinco variáveis numéricas da seleção e o boxplot de `receita_liquida`, já definido no código. Não é necessário escolher uma variável ou editar a célula.

A tabela localiza os registros correspondentes aos pontos destacados no boxplot, inclusive quando não são o mínimo ou o máximo. Observem o primeiro registro exibido e relacionem seu valor à distribuição, à loja e à data. Se a verificação não mostrar indício relevante, registrem isso.

As duas despesas com formatos mistos permanecem com a inspeção quantitativa pendente. Um ponto destacado no gráfico é um indício a examinar; não comprova erro e não determina tratamento.

```python
# Resumo e boxplot definidos previamente; nenhum dado é alterado.
variaveis_numericas = [
    "entradas", "espera_media_min", "receita_liquida",
    "desconto_campanha_pct", "disponibilidade_produtos_pct"
]
print("Resumo das variáveis numéricas; count indica os valores não ausentes usados:")
display(dados[variaveis_numericas].describe().T[["count", "min", "25%", "50%", "75%", "max"]])
print("Inspeção quantitativa pendente para as duas despesas com formatos mistos.")

variavel_observada = "receita_liquida"
serie = dados[variavel_observada].dropna()
fig, ax = plt.subplots(figsize=(4, 6))
elementos = ax.boxplot(serie, vert=True)
ax.set_ylabel("Receita líquida (R$)")
ax.set_xticks([])
ax.set_title("Boxplot — receita líquida")
fig.tight_layout()
plt.show()

# Localizar os valores já destacados pelo gráfico, sem aplicar tratamento.
valores_destacados = elementos["fliers"][0].get_ydata()
campos = ["id_registro", "data", "loja", variavel_observada]
registros_destacados = dados.loc[dados[variavel_observada].isin(valores_destacados), campos]
registros_destacados = registros_destacados.sort_values(variavel_observada, ascending=False)
print("Registros correspondentes aos pontos destacados no boxplot:", len(registros_destacados))
if registros_destacados.empty:
    print("O boxplot não destaca pontos separados. Registrem o resultado desta verificação.")
else:
    print("Até 20 registros, do maior para o menor valor. Examinem o primeiro exibido:")
    display(registros_destacados.head(20))
plt.close(fig)
```
