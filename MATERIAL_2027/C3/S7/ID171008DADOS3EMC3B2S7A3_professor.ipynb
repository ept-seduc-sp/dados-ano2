## 2. Evitem contar o mesmo dia duas vezes — 8 minutos

A comparação entre lojas depende de registros que representem as operações ocorridas. Executem a verificação da duplicação integral e a criação da base de trabalho. Observem qual registro aparece repetido e as contagens antes e depois.

```python
# Comparação integral das 15 colunas; nenhuma coluna é escolhida pelo estudante.
mascara_copias = dados_originais.duplicated(keep=False)
envolvidos = dados_originais.loc[mascara_copias]
print("Ocorrências envolvidas na duplicação integral:")
display(envolvidos)
copias_adicionais = int(dados_originais.duplicated().sum())

# Criar uma base de trabalho e manter uma ocorrência de cada registro completo.
dados_preparados = dados_originais.drop_duplicates(keep="first").copy()
ids_envolvidos = envolvidos["id_registro"].unique()
conferencia_copias = pd.DataFrame({
    "id_registro": ids_envolvidos,
    "ocorrencias_antes": [int(dados_originais["id_registro"].eq(i).sum()) for i in ids_envolvidos],
    "ocorrencias_depois": [int(dados_preparados["id_registro"].eq(i).sum()) for i in ids_envolvidos]
})
display(conferencia_copias)
display(pd.DataFrame({
    "verificacao": ["Linhas na base original", "Cópias integrais adicionais retiradas",
                    "Linhas na base principal", "Cópias adicionais restantes"],
    "quantidade": [len(dados_originais), copias_adicionais, len(dados_preparados),
                   int(dados_preparados.duplicated().sum())]
}))
```

## 3. Preparem indicadores de desempenho — 10 minutos

O total de cupons e o lucro bruto, isoladamente, não mostram sua relação com o movimento e a receita. Executem a conferência dos denominadores e os cálculos por loja-dia: cupons por 100 entradas = 100 × cupons_emitidos ÷ entradas; margem bruta (%) = 100 × lucro_bruto ÷ receita_liquida. Observem o exemplo R00001.

Antes dos cálculos, o código confere as ausências e os valores não positivos de `entradas` e `receita_liquida`, que são os denominadores. Também confere a disponibilidade dos numeradores. O cálculo usa os valores completos; somente a exibição é arredondada para duas casas decimais. Os resultados são calculados por loja-dia, sem agregar lojas ou datas.

```python
denominadores = ["entradas", "receita_liquida"]
conferencia_denominadores = pd.DataFrame({
    "denominador": denominadores,
    "ausentes": [int(dados_preparados[c].isna().sum()) for c in denominadores],
    "nao_positivos": [int(dados_preparados[c].le(0).sum()) for c in denominadores]
})
print("Conferência dos denominadores:")
display(conferencia_denominadores)
print("Numeradores ausentes:")
display(dados_preparados[["cupons_emitidos", "lucro_bruto"]].isna().sum().rename("ausentes").to_frame())

# Reaplicação dos indicadores conhecidos; colunas originais preservadas.
entradas_validas = dados_preparados["entradas"].where(dados_preparados["entradas"].gt(0))
receitas_validas = dados_preparados["receita_liquida"].where(dados_preparados["receita_liquida"].gt(0))
dados_preparados["cupons_por_100_entradas"] = 100 * dados_preparados["cupons_emitidos"] / entradas_validas
dados_preparados["margem_bruta_pct"] = 100 * dados_preparados["lucro_bruto"] / receitas_validas
indicadores = ["cupons_por_100_entradas", "margem_bruta_pct"]
print("Quantidade de valores calculados:")
display(dados_preparados[indicadores].count().rename("valores_calculados").to_frame())
print("Conferência do registro R00001 (ponto indica separador decimal):")
exemplo = dados_preparados.loc[dados_preparados["id_registro"].eq("R00001")]
display(exemplo[["id_registro", "entradas", "cupons_emitidos", "cupons_por_100_entradas"]].round(2))
display(exemplo[["id_registro", "receita_liquida", "lucro_bruto", "margem_bruta_pct"]].round(2))
print("Colunas originais:", dados_originais.shape[1], "| Colunas na base principal:", dados_preparados.shape[1])
```
