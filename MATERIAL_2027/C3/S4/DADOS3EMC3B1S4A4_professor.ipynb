## 2. Como ler as variáveis

- `mes` — mês de referência, de dezembro de 2024 a dezembro de 2025 (13 meses).
- `ordem_mes` — número de 1 a 13, usado apenas para manter os meses na ordem correta.
- `atividade` — atividade do comércio varejista.
- `indice_volume` — índice de base fixa do volume de vendas.

**Atenção à unidade.** O índice **não é porcentagem de variação**. Na Tabela 5 usada nesta
atividade, a base é **2014 = 100**. Um valor de 170 indica volume 70% acima dessa referência;
um valor de 95 indica volume 5% abaixo dela.

A atividade focaliza duas das nove séries do conjunto:

- **Tecidos, vestuário e calçados** — chamada de "vestuário" ao longo do roteiro.
- **Informática e comunicação** — chamada de "informática" ao longo do roteiro.


```python
# Separa as duas atividades analisadas nesta aula.
vestuario = dados[dados["atividade"] == "Tecidos, vestuário e calçados"]
informatica = dados[dados["atividade"] == "Informática e comunicação"]

print("Linhas de vestuário:", len(vestuario))
print("Linhas de informática:", len(informatica))
```

## Etapa 1 — Executar e observar

Execute as três células a seguir. Cada uma gera uma das visualizações que o gestor recebeu.
Nesta etapa você apenas **observa**: ainda não é hora de concluir nada.


```python
# GRÁFICO 1 — linha: o índice das duas atividades ao longo dos 13 meses.
plt.figure(figsize=(11, 4.5))

for atividade in ["Tecidos, vestuário e calçados", "Informática e comunicação"]:
    serie = dados[dados["atividade"] == atividade]
    plt.plot(serie["mes"], serie["indice_volume"], marker="o", label=atividade)

plt.xlabel("Mês")
plt.ylabel("Índice de volume de vendas (2014 = 100)")
plt.title("Gráfico 1 — Volume de vendas mês a mês")
plt.xticks(rotation=40, ha="right")
plt.legend()
plt.grid(axis="y", linewidth=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRÁFICO 2 — barras: as oito atividades do varejo em dezembro de 2025.
dez = dados[(dados["mes"] == "dez/2025") & (dados["atividade"] != "Comércio varejista (total)")]

plt.figure(figsize=(10, 4.5))
plt.bar(dez["atividade"], dez["indice_volume"])

plt.xlabel("Atividade")
plt.ylabel("Índice de volume de vendas (2014 = 100)")
plt.title("Gráfico 2 — Volume de vendas por atividade em dez/2025")
plt.xticks(rotation=35, ha="right")
plt.grid(axis="y", linewidth=0.3)
plt.tight_layout()
plt.show()
```

```python
# GRÁFICO 3 — boxplot: a distribuição dos 13 índices de cada atividade.
plt.figure(figsize=(7, 4.5))
plt.boxplot([vestuario["indice_volume"], informatica["indice_volume"]],
            tick_labels=["Vestuário", "Informática"])

plt.ylabel("Índice de volume de vendas (2014 = 100)")
plt.title("Gráfico 3 — Distribuição dos 13 índices mensais")
plt.grid(axis="y", linewidth=0.3)
plt.tight_layout()
plt.show()
```

## Etapa 4 — Reconhecer possíveis exceções

No **Gráfico 3**, as duas atividades apresentam pontos desenhados **separadamente, acima da
haste superior** da caixa. São valores que se afastam do restante da distribuição.

> **Atenção.** Um ponto pode representar mais de um mês, quando dois meses têm o mesmo valor.
> Por isso não conte os pontos do gráfico: use a célula abaixo para descobrir **quais meses**
> são esses.

A célula calcula o limite superior do boxplot para cada atividade e lista os meses acima dele.


> **Atenção à lacuna da Etapa 4.** Neste notebook do professor, a célula também permanece intencionalmente com `mes_anterior = "____"`, para espelhar a tarefa entregue aos estudantes. Isso não é erro do arquivo. A resposta esperada é `"dez/2024"`. Para testar previamente a comparação completa, preencha temporariamente essa lacuna e, para manter o arquivo-modelo, restaure-a depois.

```python
def meses_afastados(tabela):
    """Devolve os meses cujo índice fica acima da haste superior do boxplot."""
    q1 = tabela["indice_volume"].quantile(0.25)
    q3 = tabela["indice_volume"].quantile(0.75)
    limite_superior = q3 + 1.5 * (q3 - q1)
    acima = tabela[tabela["indice_volume"] > limite_superior]
    return round(limite_superior, 1), acima[["mes", "indice_volume"]]

limite_v, afastados_v = meses_afastados(vestuario)
limite_i, afastados_i = meses_afastados(informatica)

print("VESTUÁRIO — limite superior:", limite_v)
print(afastados_v.to_string(index=False))
print()
print("INFORMÁTICA — limite superior:", limite_i)
print(afastados_i.to_string(index=False))
```
