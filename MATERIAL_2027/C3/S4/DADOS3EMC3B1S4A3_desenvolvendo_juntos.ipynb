## Etapa 1 — Comparar atividades

**Pergunta da análise**

**Em dezembro de 2025, como as oito atividades se comparam quanto à variação percentual do volume de vendas em relação a novembro?**

Antes de executar o código, identifique:

- a variável que representa as atividades;
- a variável que mede a variação do volume de vendas em dezembro;
- o tipo de gráfico adequado para comparar as oito categorias.

Nesta primeira etapa, o professor demonstrará o código-base. Execute a célula e confira se a visualização responde exatamente à pergunta.

```python
plt.figure(figsize=(10, 5))

plt.bar(dados["atividade"], dados["volume_dez"])

plt.axhline(0, linewidth=1)
plt.xlabel("Atividade")
plt.ylabel("Variação do volume de vendas (%)")
plt.title("Variação do volume de vendas em dezembro")
plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.show()
```

**Compare a tabela com o gráfico**

1. Qual atividade apresenta a maior variação em dezembro? E a menor?
2. Essa resposta já podia ser obtida na tabela?
3. O que o gráfico tornou **mais rápido ou mais evidente** ao comparar as oito atividades?

**Registre uma conclusão curta sobre o que a visualização acrescentou à leitura dos valores.**

## Etapa 2 — Acompanhar uma mudança no tempo

**Pergunta da análise**

**Como variou, mês a mês, o volume de vendas de Informática e comunicação entre outubro, novembro e dezembro de 2025?**

A sequência deve ser lida exatamente nesta ordem:

**outubro → novembro → dezembro**

O código abaixo executa, mas a representação ainda não é a mais adequada para acompanhar essa sequência temporal.

1. Identifique a representação mais adequada.
2. Modifique **somente a linha indicada**.
3. Execute a célula.
4. Confira se a visualização permite perceber claramente a mudança entre os três meses.

```python
informatica = dados[dados["atividade"] == "Informática e comunicação"].iloc[0]

meses = ["Out", "Nov", "Dez"]
valores = [informatica["volume_out"], informatica["volume_nov"], informatica["volume_dez"]]

plt.figure(figsize=(8, 4))

# MODIFIQUE ESTA LINHA para a representação mais adequada.
plt.bar(meses, valores)

plt.xlabel("Mês")
plt.ylabel("Variação do volume de vendas (%)")
plt.title("Informática e comunicação — outubro a dezembro")

plt.tight_layout()
plt.show()
```

**Compare a tabela com o gráfico**

1. De outubro para novembro, a variação aumentou ou diminuiu?
2. De novembro para dezembro, aumentou ou diminuiu?
3. O que a visualização torna mais evidente sobre essa sequência do que a leitura isolada dos três valores?

**Registre a trajetória em uma frase, citando os três meses.**

## Etapa 3 — Investigar a relação entre duas medidas

**Pergunta da análise**

**Em dezembro de 2025, as atividades com maior variação percentual do volume de vendas também tendem a apresentar maior variação percentual da receita nominal?**

Para responder, precisamos observar **as mesmas oito atividades** usando duas medidas do mesmo mês:

- `volume_dez` — variação percentual do volume de vendas;
- `receita_dez` — variação percentual da receita nominal.

O código abaixo executa, mas contém **duas escolhas que precisam ser revistas**:

- uma das variáveis;
- o tipo de representação.

Modifique apenas os trechos indicados, execute e confira se a visualização permite investigar a relação entre as duas medidas.

```python
x = dados["volume_dez"]

# MODIFIQUE: qual variável deve ser comparada com volume_dez?
y = dados["volume_nov"]

plt.figure(figsize=(7, 5))

# MODIFIQUE: escolha a representação adequada para investigar a relação entre x e y.
plt.plot(x, y, marker="o")

plt.xlabel("Variação do volume de vendas em dezembro (%)")
plt.ylabel("Variação da receita nominal em dezembro (%)")
plt.title("Volume de vendas × receita nominal")

plt.tight_layout()
plt.show()
```

**Compare a tabela com o gráfico**

1. A maior parte dos pontos sugere que as duas medidas variam na mesma direção?
2. Existe alguma atividade que se afasta desse comportamento geral?
