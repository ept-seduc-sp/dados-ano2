## 2. Como ler as variáveis

- `id_pedido` — identificador anônimo do pedido.
- `peso_produto_g` — peso do produto, em gramas.
- `distancia_aproximada_km` — distância aproximada, em quilômetros, entre vendedor e cliente. É uma distância em linha reta obtida a partir das coordenadas associadas aos prefixos de CEP; **não representa a rota real percorrida**.
- `tempo_entrega_dias` — número de dias entre a compra e a entrega ao cliente.
- `valor_frete` — valor do frete, em reais.
- `preco_produto` — preço do produto, em reais.
- `nota_avaliacao` — avaliação do cliente, em escala ordenada de 1 a 5.

## Etapa 1 — Produzir as evidências

Execute as três células a seguir. Cada célula gera um gráfico de dispersão e calcula o coeficiente de correlação `r`.

Observe primeiro a **organização dos pontos** e depois compare com o valor de `r`.

```python
# GRÁFICO 1 — peso do produto × valor do frete
r_peso = dados["peso_produto_g"].corr(dados["valor_frete"])

plt.figure(figsize=(9, 5))
plt.scatter(dados["peso_produto_g"], dados["valor_frete"], s=14, alpha=0.22)
plt.xlabel("Peso do produto (g)")
plt.ylabel("Valor do frete (R$)")
plt.title("Peso do produto × valor do frete")
plt.grid(linewidth=0.3)
plt.tight_layout()
plt.show()

print(f"r = {r_peso:+.3f}")
```

```python
# GRÁFICO 2 — distância aproximada × valor do frete
r_distancia = dados["distancia_aproximada_km"].corr(dados["valor_frete"])

plt.figure(figsize=(9, 5))
plt.scatter(dados["distancia_aproximada_km"], dados["valor_frete"], s=14, alpha=0.22)
plt.xlabel("Distância aproximada entre vendedor e cliente (km)")
plt.ylabel("Valor do frete (R$)")
plt.title("Distância aproximada × valor do frete")
plt.grid(linewidth=0.3)
plt.tight_layout()
plt.show()

print(f"r = {r_distancia:+.3f}")
```

```python
# GRÁFICO 3 — tempo de entrega × valor do frete
r_tempo = dados["tempo_entrega_dias"].corr(dados["valor_frete"])

plt.figure(figsize=(9, 5))
plt.scatter(dados["tempo_entrega_dias"], dados["valor_frete"], s=14, alpha=0.22)
plt.xlabel("Tempo de entrega (dias)")
plt.ylabel("Valor do frete (R$)")
plt.title("Tempo de entrega × valor do frete")
plt.grid(linewidth=0.3)
plt.tight_layout()
plt.show()

print(f"r = {r_tempo:+.3f}")
```

```python
# Resumo dos indicadores das três relações
resumo = pd.DataFrame({
    "Relação": [
        "Peso do produto × valor do frete",
        "Distância aproximada × valor do frete",
        "Tempo de entrega × valor do frete",
    ],
    "r": [r_peso, r_distancia, r_tempo],
})

resumo.assign(r=resumo["r"].round(3))
```

**2. Justificativa:** a ordem deve combinar os gráficos com a magnitude dos coeficientes: `0,603 > 0,407 > 0,221`. A nuvem de peso × frete apresenta uma tendência crescente mais organizada; distância × frete é mais espalhada; tempo de entrega × frete apresenta o maior espalhamento entre as três.

### Registro 3 — expectativa

### Registro 4 — decisão

A análise adicional usa Spearman porque `nota_avaliacao` é ordinal.

| Relação adicional | `rₛ` |
|---|---:|
| Tempo de entrega × nota de avaliação | **−0,261** |
| Preço do produto × nota de avaliação | **−0,016** |

**4. Variável a priorizar:** **tempo de entrega** para investigação e possível ação. Sua associação com a nota é claramente maior em magnitude do que a associação observada para preço.

**5. Direção coerente:** **reduzir o tempo de entrega**. Como a relação é negativa, tempos menores estão associados a avaliações maiores no recorte. Para preço, `rₛ` está próximo de zero e não sustenta uma direção de mudança visando avaliações melhores.

**6. Cuidado:** a correlação não demonstra que reduzir o tempo de entrega **causará** aumento das notas. A relação orienta uma hipótese e uma prioridade de investigação/ação que deve ser acompanhada e avaliada.

### Registro 5 — conclusão

Exemplo de resposta adequada:

> No recorte analisado, peso do produto apresenta a relação linear mais intensa com o valor do frete (`r = +0,603`), e seu gráfico mostra uma tendência crescente mais organizada que as demais. Distância também se associa positivamente ao frete (`r = +0,407`), enquanto tempo de entrega apresenta a menor intensidade entre as três (`r = +0,221`). Essas relações ajudam a formar expectativas sobre o frete, mas não determinam seu valor nem comprovam causalidade.

Aceite formulações equivalentes que integrem evidência visual e numérica e mantenham a conclusão limitada ao que os dados sustentam.

```python
# Conferência do professor — coeficientes usados na questão de decisão.
# Spearman pode ser obtido como a correlação de Pearson entre os postos.
rs_tempo_nota = dados["tempo_entrega_dias"].rank().corr(dados["nota_avaliacao"].rank())
rs_preco_nota = dados["preco_produto"].rank().corr(dados["nota_avaliacao"].rank())

print(f"Tempo de entrega × nota de avaliação: r_s = {rs_tempo_nota:+.3f}")
print(f"Preço do produto × nota de avaliação: r_s = {rs_preco_nota:+.3f}")
```
