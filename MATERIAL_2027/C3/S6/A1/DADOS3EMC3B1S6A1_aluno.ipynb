## Etapa 2 — Verifiquem pontos de atenção

Antes de escolher uma pergunta para o projeto, verifiquem se há registros repetidos ou informações ausentes.

Executem a célula abaixo e examinem as saídas.

```python
# Verificação de registros repetidos
copias_duplicadas = dados.duplicated().sum()
print("Cópias duplicadas encontradas:", copias_duplicadas)

if copias_duplicadas > 0:
    print("\nRegistros envolvidos na duplicação:")
    display(dados[dados.duplicated(keep=False)].sort_values(["id_registro", "data"]))

# Verificação de valores ausentes
ausentes = dados.isna().sum()
ausentes = ausentes[ausentes > 0].to_frame("valores_ausentes")

print("\nColunas com valores ausentes:")
display(ausentes)
```
