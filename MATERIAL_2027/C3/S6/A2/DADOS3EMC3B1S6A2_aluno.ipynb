## Etapa 3 — Preparem os indicadores necessários

Algumas perguntas exigem combinar informações já presentes no conjunto.

### Indicador 1 — Cupons por 100 entradas

$$
\text{cupons por 100 entradas} = \frac{\text{cupons emitidos}}{\text{entradas}} \times 100
$$

Esse indicador representa a **quantidade de transações registradas para cada 100 entradas**, calculada para uma mesma loja em um mesmo dia.

> Como os dados não identificam pessoas únicas, esse indicador **não representa a porcentagem de clientes distintos que compraram**.

### Indicador 2 — Margem bruta percentual

$$
\text{margem bruta (\%)} = \frac{\text{lucro bruto}}{\text{receita líquida}} \times 100
$$

Esse indicador representa a **parcela da receita líquida que corresponde ao lucro bruto de uma loja em um dia**.

Exemplo: uma margem bruta de 30% corresponde a R$ 30 de lucro bruto para cada R$ 100 de receita líquida.

> A margem bruta **não representa o lucro final da loja depois de todas as despesas**. Neste projeto, o lucro bruto é usado como medida de resultado comercial porque o conjunto não contém as despesas gerais necessárias para calcular o lucro líquido.

Executem a célula abaixo. O código confere os denominadores e cria os dois indicadores para cada registro loja-dia.

```python
# Conferência dos denominadores usados nos indicadores
entradas_ausentes = dados_trabalho["entradas"].isna().sum()
receitas_ausentes = dados_trabalho["receita_liquida"].isna().sum()

entradas_nao_positivas = (
    dados_trabalho["entradas"].notna() & (dados_trabalho["entradas"] <= 0)
).sum()
receitas_nao_positivas = (
    dados_trabalho["receita_liquida"].notna() & (dados_trabalho["receita_liquida"] <= 0)
).sum()

print("Registros com entradas ausentes:", entradas_ausentes)
print("Registros com entradas <= 0:", entradas_nao_positivas)
print("Registros com receita_liquida ausente:", receitas_ausentes)
print("Registros com receita_liquida <= 0:", receitas_nao_positivas)

if any([
    entradas_ausentes, receitas_ausentes,
    entradas_nao_positivas, receitas_nao_positivas
]):
    raise ValueError("Há denominadores ausentes ou não positivos para o cálculo dos indicadores.")

# Indicadores calculados para cada loja em cada dia
dados_trabalho["cupons_por_100_entradas"] = (
    dados_trabalho["cupons_emitidos"] / dados_trabalho["entradas"] * 100
)

dados_trabalho["margem_bruta_pct"] = (
    dados_trabalho["lucro_bruto"] / dados_trabalho["receita_liquida"] * 100
)

colunas_exemplo = [
    "data", "loja", "entradas", "cupons_emitidos", "cupons_por_100_entradas",
    "receita_liquida", "lucro_bruto", "margem_bruta_pct"
]

display(dados_trabalho[colunas_exemplo].head().round(2))
```

## Etapa 4 — Construam o plano de investigação

Agora, determinem **como** cada parte do problema será investigada. Usem somente procedimentos já trabalhados no bimestre.

### Banco de procedimentos

**P1 — Caracterização:** medidas de tendência central e dispersão, acompanhadas de uma representação da distribuição, como histograma ou boxplot.

**P2 — Associação entre variáveis quantitativas:** diagrama de dispersão + coeficiente de correlação `r`, com interpretação da direção, intensidade e espalhamento dos pontos.

**P3 — Comparação entre grupos ou níveis:** tabela-resumo + representação que permita comparar distribuições, como boxplot.

**P4 — Relação global e comparação contextualizada:** primeiro examinar a relação no conjunto de registros e, depois, comparar registros de contextos mais semelhantes para verificar se a visão global resume adequadamente os dados.

### Evidências que podem ser procuradas

**E1 —** valores típicos, variabilidade, forma da distribuição e registros que mereçam atenção.  
**E2 —** direção e intensidade da associação linear e espalhamento dos pontos.  
**E3 —** associações entre condições registradas e diferenças observadas no aproveitamento do movimento, incluindo se a relação global permanece adequada em contextos comparáveis.  
**E4 —** comportamento do aproveitamento, do lucro bruto e da margem bruta entre níveis de desconto e se uma medida global resume adequadamente os grupos.
