# Situação 1 — Cidades sem hierarquia

Uma empresa atende ordens de serviço em cidades diferentes.

A análise precisa representar `cidade_atendimento` numericamente, **mantendo cada cidade distinta e sem estabelecer uma ordem entre elas**.


```python
situacao_1 = pd.DataFrame({
    "ordem": ["OS01", "OS02", "OS03", "OS04"],
    "cidade_atendimento": ["Campinas", "Sorocaba", "Jundiaí", "Campinas"],
})

display(situacao_1)
```

Quando tiver registrado sua decisão, execute a próxima célula. Ela não depende da sua resposta.


```python
situacao_1_transformada = pd.get_dummies(
    situacao_1,
    columns=["cidade_atendimento"],
    prefix="cidade",
    dtype=int,
)

display(situacao_1_transformada)
```

# Situação 2 — Reorganizando faixas

A empresa registra `tempo_reparo` nas quatro categorias abaixo. Para uma análise específica, interessa comparar apenas:

**reparos de até 3 horas × reparos com mais de 3 horas**.


```python
situacao_2 = pd.DataFrame({
    "registro": ["R01", "R02", "R03", "R04"],
    "tempo_reparo": [
        "Até 1 hora",
        "Mais de 1 até 3 horas",
        "Mais de 3 até 6 horas",
        "Mais de 6 horas",
    ],
})

display(situacao_2)
```

### Decida antes de executar

No caderno:

1. Proponha a **nova categorização** necessária para a comparação.
2. Indique em qual nova categoria ficará cada faixa original.
3. Classifique a transformação como **representação computacional** ou **analítica**.

Só depois execute a próxima célula.


```python
mapa_tempo = {
    "Até 1 hora": "Até 3 horas",
    "Mais de 1 até 3 horas": "Até 3 horas",
    "Mais de 3 até 6 horas": "Mais de 3 horas",
    "Mais de 6 horas": "Mais de 3 horas",
}

situacao_2_transformada = situacao_2.copy()
situacao_2_transformada["tempo_reparo_agrupado"] = (
    situacao_2_transformada["tempo_reparo"].map(mapa_tempo)
)

display(situacao_2_transformada)
```

### Compare e interprete

- **Que detalhe da informação original deixa de ser distinguido** depois do agrupamento?

# Situação 3 — O conteúdo é uma data, mas o tipo ainda é texto

Os registros de `data_abertura` estão consistentes, mas o sistema ainda os reconhece como texto.

Para o uso previsto, esses valores precisam ser reconhecidos computacionalmente como **datas**.

> Nesta situação, não faremos análise temporal. O foco é somente a transformação do tipo da variável.


```python
situacao_3 = pd.DataFrame({
    "ordem": ["OS21", "OS22", "OS23"],
    "data_abertura": ["2026-10-03", "2026-10-07", "2026-10-11"],
})

print("Tipo antes da transformação:", situacao_3["data_abertura"].dtype)
display(situacao_3)
```

### Decida antes de executar

No caderno, registre:

1. Qual transformação é necessária?
2. Qual deve ser o **tipo antes** e o **tipo depois**?
3. O que você espera que seja preservado?
4. A transformação é de **representação computacional** ou **analítica**?

Depois, execute a próxima célula.


```python
situacao_3_transformada = situacao_3.copy()
situacao_3_transformada["data_abertura"] = pd.to_datetime(
    situacao_3_transformada["data_abertura"],
    format="%Y-%m-%d",
)

print("Tipo depois da transformação:", situacao_3_transformada["data_abertura"].dtype)
display(situacao_3_transformada)
```

### Compare e interprete

No caderno, responda:

- O significado das datas mudou?
- O que mudou na forma como o sistema reconhece a variável?

# Situação 4 — Construindo uma informação que ainda não existe

A empresa quer comparar **quantas ordens foram concluídas por hora de trabalho da equipe**.

A tabela possui `ordens_concluidas` e `horas_equipe`, mas essa informação ainda não aparece diretamente.


```python
situacao_4 = pd.DataFrame({
    "equipe": ["A", "B", "C"],
    "ordens_concluidas": [12, 15, 20],
    "horas_equipe": [4, 6, 5],
})

display(situacao_4)
```

### Decida antes de executar

No caderno:

1. Que nova variável você criaria para atender à necessidade?
2. Escreva como ela pode ser calculada a partir das variáveis existentes.
3. Classifique a transformação como **representação computacional** ou **analítica**.
4. Antes de executar, estime qual equipe deverá ter o maior valor da nova variável.

A próxima célula executará o cálculo independentemente da sua resposta.


```python
situacao_4_transformada = situacao_4.copy()
situacao_4_transformada["ordens_por_hora_equipe"] = (
    situacao_4_transformada["ordens_concluidas"]
    / situacao_4_transformada["horas_equipe"]
).round(2)

display(situacao_4_transformada)
```
