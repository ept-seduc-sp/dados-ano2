## Como ler os dados

O arquivo `DADOS3EMC3B2S10A4_CP_PMC.csv` contém 36 registros mensais.

- `data` — mês de referência.
- `indice_volume_vendas` — índice do volume de vendas da atividade **Móveis e eletrodomésticos**.

O índice usa como referência a média mensal de 2022, igual a 100. Assim, valores acima de 100 indicam nível de vendas acima dessa referência; valores abaixo de 100 indicam nível abaixo dela.

**Importante:** o índice representa o comportamento do **setor**, e não o faturamento ou o número de unidades vendidas de uma empresa específica.

## 1. Conheça a série

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from statsmodels.tsa.seasonal import seasonal_decompose

arquivo = 'DADOS3EMC3B2S10A4_CP_PMC.csv'
dados = pd.read_csv(arquivo, parse_dates=['data'])
dados['ano'] = dados['data'].dt.year
dados['mes_num'] = dados['data'].dt.month

print(f'Número de registros: {len(dados)}')
print(f'Período: {dados["data"].min():%m/%Y} a {dados["data"].max():%m/%Y}')
display(dados[['data','indice_volume_vendas']])
```

## 3. Compare os anos e procure padrões recorrentes

```python
meses = ['jan','fev','mar','abr','mai','jun','jul','ago','set','out','nov','dez']

fig, ax = plt.subplots(figsize=(13.33, 7.5))
for ano, grupo in dados.groupby('ano'):
    grupo = grupo.sort_values('mes_num')
    ax.plot(grupo['mes_num'], grupo['indice_volume_vendas'], marker='o', linewidth=2.2, label=str(ano))

ax.set_title('Comparação dos meses em 2023, 2024 e 2025', fontsize=22)
ax.set_xlabel('Mês do ano', fontsize=16)
ax.set_ylabel('Índice do volume de vendas', fontsize=16)
ax.set_xticks(range(1,13), meses)
ax.tick_params(axis='both', labelsize=11)
ax.legend(title='Ano')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

## 5. Use a decomposição como apoio à análise

Agora separe visualmente a série em **tendência, sazonalidade e ruído (resíduo)**. O código já está preparado: o objetivo é interpretar os componentes, não desenvolver o método de decomposição.

```python
serie = pd.Series(dados['indice_volume_vendas'].to_numpy(), index=dados['data'])

decomp = seasonal_decompose(
    serie,
    model='additive',
    period=12,
    extrapolate_trend='freq'
)

componentes = [
    ('Série observada', decomp.observed),
    ('Tendência', decomp.trend),
    ('Sazonalidade', decomp.seasonal),
    ('Ruído (resíduo)', decomp.resid),
]

fig, eixos = plt.subplots(4, 1, figsize=(13.33, 10), sharex=True)
fig.suptitle('Decomposição da série de móveis e eletrodomésticos', fontsize=20)

for eixo, (rotulo, componente) in zip(eixos, componentes):
    eixo.plot(componente.index, componente.values, linewidth=2)
    eixo.set_ylabel(rotulo, fontsize=11)
    eixo.tick_params(axis='both', labelsize=9)
    eixo.spines['top'].set_visible(False)
    eixo.spines['right'].set_visible(False)

eixos[-1].set_xlabel('Tempo', fontsize=12)
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
