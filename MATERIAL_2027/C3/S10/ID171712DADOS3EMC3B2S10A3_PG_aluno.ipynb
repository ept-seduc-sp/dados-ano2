## 1. Conheça e visualize a série

Use o arquivo `DADOS3EMC3B2S10A3_PG_airpassengers.csv`.


```python
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

dados = pd.read_csv('DADOS3EMC3B2S10A3_PG_airpassengers.csv', parse_dates=['data'])
serie = pd.Series(dados['passageiros_mil'].to_numpy(), index=dados['data'])

print(f'Observações: {len(dados)}')
print(f'Período: {dados["data"].min():%b/%Y} a {dados["data"].max():%b/%Y}')
display(dados.head())

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(serie.index, serie.values, linewidth=2)
ax.set_title('Passageiros internacionais por mês — 1949 a 1960')
ax.set_xlabel('Tempo')
ax.set_ylabel('Passageiros (mil)')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

### Observe antes da decomposição

1. Que regularidades você percebe na série?
2. Como a tendência se comporta?
3. O padrão sazonal parece regular?


## 2. Decomponha a série

**Série observada = Tendência + Sazonalidade + Resíduo**


```python
decomp = seasonal_decompose(
    serie,
    model='additive',
    period=12,
    extrapolate_trend='freq'
)

componentes = [
    ('Série observada', decomp.observed),
    ('Tendência', decomp.trend),
    ('Sazonalidade', decomp.seasonal),
    ('Ruído (resíduo)', decomp.resid),
]

fig, eixos = plt.subplots(4, 1, figsize=(12, 10), sharex=True)
fig.suptitle('Decomposição — AirPassengers', fontsize=16)

for eixo, (rotulo, componente) in zip(eixos, componentes):
    eixo.plot(componente.index, componente.values, linewidth=1.8)
    eixo.set_ylabel(rotulo)
    eixo.spines['top'].set_visible(False)
    eixo.spines['right'].set_visible(False)

eixos[-1].set_xlabel('Tempo')
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
