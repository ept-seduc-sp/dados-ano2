## Orientações
O notebook permite conferir a montagem dos vetores e a execução posição por posição das operações `F = A + B` e `P = F - C`. Os resultados devem ser interpretados em relação à ordem das plataformas e ao critério da atividade.

### Etapa 1 — Vetores A, B e C

```python
A = [20, 63, 6, 47, 78, 38, 4]
B = [8, 19, 3, 12, 13, 22, 3]
C = [52, 9, 77, 25, 4, 17, 79]

print("A =", A)
print("B =", B)
print("C =", C)
```

### Etapa 2 — Calcular F e P

```python
F = [A[i] + B[i] for i in range(len(A))]
P = [F[i] - C[i] for i in range(len(F))]

print("F =", F)
print("P =", P)
```

**Saídas esperadas:**  
`F = [28, 82, 9, 59, 91, 60, 7]`  
`P = [-24, 73, -68, 34, 87, 43, -72]`

### Etapa 3 — Relacionar P às plataformas

```python
plataformas = ["Facebook", "Instagram", "X/Twitter", "TikTok", "WhatsApp", "YouTube", "Discord"]

for i in range(len(plataformas)):
    print(plataformas[i], "→", P[i])
```

### Etapa 4 — Interpretação esperada
Os dois maiores valores são WhatsApp → 87 p.p. e Instagram → 73 p.p. Pontuações positivas indicam uso frequente maior que “Nunca ou quase nunca”; pontuações negativas indicam o contrário. A recomendação esperada é WhatsApp e Instagram, com justificativa baseada nos valores de `P`.
