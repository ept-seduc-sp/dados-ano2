## Contexto da atividade

A equipe júnior de Ciência de Dados está apoiando uma campanha escolar de orientação sobre uso seguro da internet.

A equipe trabalha com dados reais sobre a frequência de uso de plataformas digitais por adolescentes e precisa usar operações vetoriais para transformar esses dados em critérios de decisão.

A campanha terá duas frentes:

- **Alerta rápido:** recado curto para chegar aos estudantes com mais frequência.
- **Reforço semanal:** conteúdo para circular ao longo da semana.

Produto esperado do estudante:

- recomendação técnica curta para a coordenação da escola.


## Parte 1 — Preparação dos dados

A ordem das plataformas deve ser mantida em todos os vetores:

[Facebook; Instagram; X/Twitter; TikTok; WhatsApp; YouTube; Discord]

Cada posição do vetor representa uma plataforma digital. Se a ordem mudar, os cálculos passam a comparar dados de plataformas diferentes.


```python
plataformas = [
    "Facebook",
    "Instagram",
    "X/Twitter",
    "TikTok",
    "WhatsApp",
    "YouTube",
    "Discord"
]

# A = várias vezes ao dia
A = [20, 63, 6, 47, 78, 38, 4]

# B = todos ou quase todos os dias
B = [8, 19, 3, 12, 13, 22, 3]

# S = pelo menos uma vez por semana
S = [12, 8, 7, 12, 4, 19, 3]

# N = nunca ou quase nunca
N = [52, 9, 77, 25, 4, 17, 79]

plataformas, A, B, S, N
```

### Conferência inicial do professor

Antes dos cálculos, confira se:

- todos os vetores têm 7 valores;
- todos seguem a mesma ordem das plataformas;
- os estudantes compreenderam que os valores representam percentuais da pesquisa;
- a atividade usa esses valores para criar critérios didáticos de decisão.


```python
# Conferência de tamanho dos vetores
tamanhos = {
    "plataformas": len(plataformas),
    "A": len(A),
    "B": len(B),
    "S": len(S),
    "N": len(N)
}

tamanhos
```

## Parte 2 — Cálculos com vetores

Critérios da prática:

D = A + B  
D representa uso frequente.

W = D + S  
W representa uso semanal ou mais.

R = W − N  
R representa a pontuação de reforço semanal.

E = 2D + S  
E representa a pontuação de alerta rápido.

Observação importante:

- R é diferença entre percentuais, por isso deve ser lido em pontos percentuais, p.p.
- E é uma pontuação criada para a atividade, não é percentual.


```python
# D = A + B
D = [a + b for a, b in zip(A, B)]
D
```

### Interpretação esperada — vetor D

D reúne as categorias “várias vezes ao dia” e “todos ou quase todos os dias”.  
Ele mostra uma leitura de uso frequente para cada plataforma, respeitando a ordem definida.

Gabarito:

D = [28; 82; 9; 59; 91; 60; 7]


```python
# W = D + S
W = [d + s for d, s in zip(D, S)]
W
```

### Interpretação esperada — vetor W

W amplia a leitura de alcance, pois soma o uso frequente com o uso pelo menos semanal.

Gabarito:

W = [40; 90; 16; 71; 95; 79; 10]


```python
# R = W - N
R = [w - n for w, n in zip(W, N)]
R
```

### Interpretação esperada — vetor R

R compara uso semanal ou mais com baixo uso.  
Como é uma diferença entre percentuais, deve ser lido em pontos percentuais, p.p.

Valores negativos indicam que o baixo uso é maior que o uso semanal ou mais naquela plataforma.

Gabarito:

R = [-12; 81; -61; 46; 91; 62; -69]


```python
# E = 2D + S
E = [2 * d + s for d, s in zip(D, S)]
E
```

### Interpretação esperada — vetor E

E é uma pontuação criada para a atividade.  
Ela dá peso maior ao uso frequente, pois multiplica D por 2 e depois acrescenta o uso semanal.

Por isso, E não deve ser lida como percentual.

Gabarito:

E = [68; 172; 25; 130; 186; 139; 17]


## Parte 3 — matriz de resultados

A matriz de resultados organiza os valores calculados para apoiar a recomendação final.

A ordem das linhas segue a lista `plataformas`.

Colunas da matriz:

- D: uso frequente
- W: uso semanal ou mais
- R: reforço semanal
- E: alerta rápido


```python
matriz_resultados = [
    [D[i], W[i], R[i], E[i]]
    for i in range(len(plataformas))
]

matriz_resultados
```

### Gabarito da matriz de resultados

| Plataforma digital | D: uso frequente | W: uso semanal ou mais | R: reforço semanal | E: alerta rápido |
|---|---:|---:|---:|---:|
| Facebook | 28 | 40 | -12 p.p. | 68 pontos |
| Instagram | 82 | 90 | 81 p.p. | 172 pontos |
| X/Twitter | 9 | 16 | -61 p.p. | 25 pontos |
| TikTok | 59 | 71 | 46 p.p. | 130 pontos |
| WhatsApp | 91 | 95 | 91 p.p. | 186 pontos |
| YouTube | 60 | 79 | 62 p.p. | 139 pontos |
| Discord | 7 | 10 | -69 p.p. | 17 pontos |


## Parte 4 — Análise e decisão

### Pergunta 1

Quais dois canais devem ser priorizados no alerta rápido?

Critério: maiores valores de E.


```python
# Dois maiores valores de E
indices_alerta = sorted(
    range(len(E)),
    key=lambda i: E[i],
    reverse=True
)[:2]

alerta_rapido = [(plataformas[i], E[i]) for i in indices_alerta]
alerta_rapido
```

### Resposta esperada — Pergunta 1

Os dois canais recomendados para o alerta rápido são:

- WhatsApp: 186 pontos
- Instagram: 172 pontos


### Pergunta 2

Qual canal deve ser usado como reforço semanal sem repetir os dois canais principais?

Critério: maior valor de R entre as plataformas que não foram escolhidas para o alerta rápido.


```python
# Maior valor de R entre as plataformas que não foram escolhidas para o alerta rápido
indices_restantes = [
    i for i in range(len(plataformas))
    if i not in indices_alerta
]

indice_reforco = max(indices_restantes, key=lambda i: R[i])
reforco_semanal = (plataformas[indice_reforco], R[indice_reforco])

reforco_semanal
```

### Resposta esperada — Pergunta 2

O canal recomendado para o reforço semanal, sem repetir os canais principais, é:

- YouTube: 62 p.p.


### Pergunta 3

Quais plataformas exigem cuidado antes de serem escolhidas como canais principais?

Critério: R negativo.


```python
# Plataformas com R negativo
plataformas_cuidado = [
    (plataformas[i], R[i])
    for i in range(len(plataformas))
    if R[i] < 0
]

plataformas_cuidado
```

### Resposta esperada — Pergunta 3

As plataformas que exigem cuidado são:

- Facebook: -12 p.p.
- X/Twitter: -61 p.p.
- Discord: -69 p.p.

Essas plataformas apresentam R negativo, indicando que o baixo uso é maior que o uso semanal ou mais.
