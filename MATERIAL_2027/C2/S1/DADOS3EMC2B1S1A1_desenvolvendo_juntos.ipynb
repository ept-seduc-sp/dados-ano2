## Prática Guiada Passo a Passo
Cada etapa deve ser demonstrada pelo professor antes de os estudantes repetirem, garantindo a compreensão da relação entre a organização matemática dos dados e sua representação no Python.

### Etapa 1 — Representação dos vetores
Crie a lista com os nomes das plataformas e os vetores das duas categorias de frequência. Observe que a mesma ordem deve ser preservada em todas as estruturas.

```python
plataformas = ["WhatsApp", "YouTube", "Instagram", "TikTok"]
varias_vezes = [78, 38, 63, 47]
todos_os_dias = [13, 22, 19, 12]

print("Plataformas:", plataformas)
print("Várias vezes ao dia:", varias_vezes)
print("Todos os dias ou quase todos os dias:", todos_os_dias)
```

### Etapa 2 — Organização dos dados em uma matriz
Reúna os percentuais na matriz numérica `dados`. Cada lista interna representa uma plataforma, e cada posição corresponde a uma categoria de frequência.

```python
dados = [
    [78, 13],
    [38, 22],
    [63, 19],
    [47, 12]
]

print("Matriz de dados:")
for linha in dados:
    print(linha)
```

### Etapa 3 — Localização e interpretação de um valor
Use os índices para localizar o valor da terceira plataforma e da primeira categoria de frequência.

```python
indice_plataforma = 2
indice_frequencia = 0

plataforma = plataformas[indice_plataforma]
valor = dados[indice_plataforma][indice_frequencia]

print("Plataforma:", plataforma)
print("Percentual:", valor)
```

**Interpretação esperada:** o índice `2` localiza o Instagram, o índice `0` localiza a categoria “Várias vezes ao dia” e a saída `63` representa **63%** dos usuários de Internet de 15 a 17 anos no recorte da pesquisa.
