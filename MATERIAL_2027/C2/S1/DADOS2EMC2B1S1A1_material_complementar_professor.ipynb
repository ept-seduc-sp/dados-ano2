## Orientações

### Etapa 1 — Organizar os vetores e os rótulos
As plataformas selecionadas são mantidas na ordem em que aparecem no relatório.

```python
plataformas = ["Instagram", "TikTok", "WhatsApp", "YouTube"]
varias_vezes = [63, 47, 78, 38]
todos_os_dias = [19, 12, 13, 22]

print("Plataformas:", plataformas)
print("Várias vezes ao dia:", varias_vezes)
print("Todos os dias ou quase todos os dias:", todos_os_dias)
```

### Etapa 2 — Construir a matriz `dados`
Cada lista interna reúne os dois percentuais da mesma plataforma.

```python
dados = [
    [63, 19],
    [47, 12],
    [78, 13],
    [38, 22]
]

print("Matriz de dados:")
for linha_dados in dados:
    print(linha_dados)
```

### Etapa 3 — Executar e interpretar
O índice 2 localiza a terceira posição porque a contagem no Python começa em 0.

```python
linha = 2
print(plataformas[linha])
print(dados[linha])
```

**Saída esperada:** `WhatsApp` e `[78, 13]`.  
**Interpretação:** 78 representa 78% em “Várias vezes ao dia” e 13 representa 13% em “Todos os dias ou quase todos os dias”. A posição do rótulo na lista corresponde à mesma linha da matriz.
