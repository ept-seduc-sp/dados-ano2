<h2 style="text-align: center;">Resumindo pagamentos com Python</h2>

Nesta prática guiada, vamos usar os mesmos 12 registros trabalhados nos slides para calcular a média, localizar a mediana e identificar a moda.



## Dados da análise

Os valores representam o **número de parcelas** registrado em 12 pagamentos com cartão de crédito.


```python
parcelas = [8, 1, 1, 8, 2, 2, 1, 3, 6, 8, 1, 1]
print("Parcelas:", parcelas)
```

## Atividade 1 — Calculando a média com Python

**Antes de executar:** retome o cálculo realizado no slide. O Python deverá produzir o mesmo resultado.

Vamos calcular a soma dos valores, a quantidade de registros e a média.


```python
total = sum(parcelas)
quantidade = len(parcelas)
media = total / quantidade

print("Soma dos valores:", total)
print("Quantidade de registros:", quantidade)
print("Média:", media)
```

**Interprete:** a saída da média não indica que existiu um pagamento com esse número exato de parcelas. Registre no roteiro o que o valor representa no conjunto analisado.


## Atividade 2 — Localizando o centro com Python

**Antes de executar:** com 12 registros, o centro do conjunto ordenado fica entre o 6º e o 7º valores. No Python, essas posições correspondem aos índices `5` e `6`.


```python
ordenados = sorted(parcelas)
print("Dados ordenados:", ordenados)

central_1 = ordenados[5]
central_2 = ordenados[6]
mediana = (central_1 + central_2) / 2

print("6º valor:", central_1)
print("7º valor:", central_2)
print("Mediana:", mediana)
```

**Interprete:** registre no roteiro o que a mediana representa depois que os dados foram colocados em ordem.


## Atividade 3 — Identificando o valor mais frequente com Python

**Antes de executar:** antecipe qual número de parcelas deverá aparecer mais vezes.

O método `.count()` conta quantas vezes um valor aparece em uma lista.


```python
frequencia_1 = parcelas.count(1)
frequencia_2 = parcelas.count(2)
frequencia_3 = parcelas.count(3)
frequencia_6 = parcelas.count(6)
frequencia_8 = parcelas.count(8)

print("1 parcela:", frequencia_1)
print("2 parcelas:", frequencia_2)
print("3 parcelas:", frequencia_3)
print("6 parcelas:", frequencia_6)
print("8 parcelas:", frequencia_8)
```

Compare as frequências exibidas. O valor com a **maior frequência** é a moda do conjunto.


```python
moda = 1
print("Moda:", moda)
print("Frequência da moda:", parcelas.count(moda))
```

**Interprete:** registre no roteiro o que a moda informa sobre o comportamento de parcelamento observado.


## Conferência final

As três medidas resumem o **mesmo conjunto de dados**, mas usam critérios diferentes.


```python
print("Média:", media)
print("Mediana:", mediana)
print("Moda:", moda)
```

- **Média:** considera todos os valores no cálculo.  
- **Mediana:** usa a posição central dos dados ordenados.  
- **Moda:** identifica o valor de maior frequência.
