## Missão 1 — Avaliações dos clientes

```python
avaliacoes = [4, 5, 5, 5, 5, 1, 5, 5, 5, 4]

dados_ordenados = sorted(avaliacoes)
media = sum(avaliacoes) / len(avaliacoes)
mediana = (dados_ordenados[4] + dados_ordenados[5]) / 2

print("Dados ordenados:", dados_ordenados)
print("Média:", media)
print("Mediana:", mediana)

print("Frequência da nota 1:", avaliacoes.count(1))
print("Frequência da nota 2:", avaliacoes.count(2))
print("Frequência da nota 3:", avaliacoes.count(3))
print("Frequência da nota 4:", avaliacoes.count(4))
print("Frequência da nota 5:", avaliacoes.count(5))

print("Moda: 5")
```

**Interpretação esperada:**  
A moda é 5, pois essa avaliação aparece 7 vezes e é a de maior frequência no conjunto.  
Média = 4,4; mediana = 5; moda = 5.


## Missão 2 — Valores de frete

```python
fretes = [13.29, 19.93, 17.87, 12.79, 18.14, 12.69, 11.85, 70.75, 11.65, 11.40]

fretes_ordenados = sorted(fretes)
media = sum(fretes) / len(fretes)
mediana = (fretes_ordenados[4] + fretes_ordenados[5]) / 2

print("Dados ordenados:", fretes_ordenados)
print("Média:", round(media, 2))
print("Mediana:", mediana)
```

```python
# Retirada temporária do valor de R$ 70,75 apenas para investigar seu efeito.

fretes_sem_valor = [13.29, 19.93, 17.87, 12.79, 18.14, 12.69, 11.85, 11.65, 11.40]

fretes_sem_ordenados = sorted(fretes_sem_valor)
media_sem = sum(fretes_sem_valor) / len(fretes_sem_valor)
mediana_sem = fretes_sem_ordenados[4]

print("Média sem R$ 70,75:", round(media_sem, 2))
print("Mediana sem R$ 70,75:", mediana_sem)

print("Alteração da média:", round(media - media_sem, 2))
print("Alteração da mediana:", round(mediana - mediana_sem, 2))
```

**Interpretação esperada:**  
Com todos os valores, média ≈ R$ 20,04 e mediana = R$ 13,04.  
Sem R$ 70,75, média ≈ R$ 14,40 e mediana = R$ 12,79.  
Para representar um **frete típico** neste recorte, a mediana é mais informativa, pois sofreu menor influência do valor muito alto.


## Missão 3 — Preços de um novo lote

```python
precos = [58.90, 239.90, 199.00, 12.99, 199.90, 21.90, 19.90, 810.00, 145.95, 53.99]

precos_ordenados = sorted(precos)
media = sum(precos) / len(precos)
mediana = (precos_ordenados[4] + precos_ordenados[5]) / 2

print("Dados ordenados:", precos_ordenados)
print("Média:", round(media, 2))
print("Mediana:", round(mediana, 2))
```

**Interpretação esperada:**  
Média ≈ R$ 176,24 e mediana ≈ R$ 102,43.  
Embora R$ 810,00 seja muito superior aos demais valores, a demanda exige considerar **todos os preços observados**.  
Para essa finalidade específica, a média é a medida adequada.
