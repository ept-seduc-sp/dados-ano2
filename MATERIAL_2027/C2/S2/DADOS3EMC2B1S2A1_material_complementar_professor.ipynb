## Como usar este notebook

Este arquivo é a versão do professor para conferência.

- O **DOCX do aluno** conduz a leitura e concentra os registros.
- O **notebook do aluno** executa e confere os procedimentos.
- A escolha das duas finalidades deve ocorrer **somente depois da comparação dos quatro pares de percentuais**.
- A nota de análise final permanece no **DOCX do aluno**.

## Etapa 1 — Construir e visualizar a matriz B

As faixas etárias selecionadas são **13 a 14 anos** e **15 a 17 anos**.

**Código Python**

```python
finalidades = [
    "Estudar ou fazer pesquisas escolares",
    "Procurar informações",
    "Criar textos, imagens, vídeos ou códigos",
    "Conversar sobre problemas ou emoções"
]

B = [
    [69, 48, 26, 16],
    [68, 60, 32, 12]
]

print("Matriz B:")
for linha in B:
    print(linha)
```

**Conferência:**  
Linhas: 13 a 14 anos; 15 a 17 anos.  
Colunas: as quatro finalidades, na mesma ordem da tabela.

## Etapa 2 — Conferir a ordem da matriz

**Código Python**

```python
numero_linhas = len(B)
numero_colunas = len(B[0])

print("Número de linhas:", numero_linhas)
print("Número de colunas:", numero_colunas)
print("Ordem da matriz:", numero_linhas, "x", numero_colunas)
```

**Resultado esperado:** 2 linhas, 4 colunas; ordem **2 × 4**.

## Etapa 3 — Comparar as quatro finalidades

Esta etapa deve ocorrer antes da escolha das duas finalidades.

**Código Python**

```python
for j in range(4):
    print(f"Coluna {j+1} - {finalidades[j]}")
    print("  13 a 14 anos:", B[0][j], "%")
    print("  15 a 17 anos:", B[1][j], "%")
```

**Resultado esperado da comparação:**

- Estudar ou fazer pesquisas escolares: 69% e 68%.
- Procurar informações: 48% e 60%.
- Criar textos, imagens, vídeos ou códigos: 26% e 32%.
- Conversar sobre problemas ou emoções: 16% e 12%.

As duas finalidades que mais se destacam são **estudar ou fazer pesquisas escolares** e **procurar informações**.

## Etapa 4 — Consultar as entradas das duas finalidades escolhidas

Para a resposta esperada, informe as colunas **1** e **2**.

**Código Python**

```python
coluna_1 = 1
coluna_2 = 2

j1 = coluna_1 - 1
j2 = coluna_2 - 1

print("Finalidade 1:", finalidades[j1])
print(f"b(1,{coluna_1}) -> B[0][{j1}] =", B[0][j1])
print(f"b(2,{coluna_1}) -> B[1][{j1}] =", B[1][j1])
print()

print("Finalidade 2:", finalidades[j2])
print(f"b(1,{coluna_2}) -> B[0][{j2}] =", B[0][j2])
print(f"b(2,{coluna_2}) -> B[1][{j2}] =", B[1][j2])
```

**Entradas esperadas:**

- \(b_{1,1}=69\) → `B[0][0]`
- \(b_{2,1}=68\) → `B[1][0]`
- \(b_{1,2}=48\) → `B[0][1]`
- \(b_{2,2}=60\) → `B[1][1]`
