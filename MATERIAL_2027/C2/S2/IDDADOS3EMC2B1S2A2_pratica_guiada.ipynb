## Contexto

A matriz **T** reúne as participações estimadas das duas unidades. A matriz **C** apresenta os custos simulados por participação.

- **Linhas de T:** faixas etárias — 13 a 14 anos; 15 a 17 anos.
- **Colunas de T:** oficinas — estudo e pesquisa; criação de conteúdos.
- **Linhas de C:** oficinas — na mesma ordem de T.
- **Colunas de C:** materiais; recursos digitais.

Execute as células na sequência e acompanhe as orientações do professor.


## Preparação — matrizes T e C

Execute a célula para disponibilizar os dados usados na Prática Guiada.

**Código Python**


```python
T = [
    [117, 44],
    [157, 73]
]

C = [
    [10, 5],
    [20, 10]
]

print("T =", T)
print("C =", C)
```
