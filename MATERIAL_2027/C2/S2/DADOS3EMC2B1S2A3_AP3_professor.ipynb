## Contexto

Plataformas de streaming registram dados como número de visualizações, duração dos títulos e horas assistidas. Nesta atividade, esses dados serão organizados em matrizes para consolidar duas semanas, estimar o total de horas assistidas e validar o resultado.

Os dados foram extraídos do **Top 10 global da Netflix**.

- [Acesso direto à planilha oficial](https://www.netflix.com/tudum/top10/data/all-weeks-global.xlsx)
- [Contexto dos indicadores de audiência](https://about.netflix.com/pt_br/news/what-we-watched-the-second-half-of-2025)

> **Recorte:** categoria *Films (English)*; semanas encerradas em 7 e 14 de setembro de 2025; títulos *KPop Demon Hunters* e *Shrek*.

## 1. Conheça os dados

Na atividade, o título oficial **KPop Demon Hunters** será apresentado como **Guerreiras do K-Pop**.

| Semana encerrada em | Título | Visualizações (milhões) | Duração (h) | Horas assistidas (milhões) |
|---|---|---:|---:|---:|
| 7 set. 2025 | Guerreiras do K-Pop | 25,4 | 1,6667 | 42,4 |
| 7 set. 2025 | Shrek | 3,6 | 1,5000 | 5,4 |
| 14 set. 2025 | Guerreiras do K-Pop | 22,6 | 1,6667 | 37,7 |
| 14 set. 2025 | Shrek | 2,9 | 1,5000 | 4,4 |

**Ordem dos títulos em todas as matrizes:**  
1. Guerreiras do K-Pop  
2. Shrek

> Visualizações e horas assistidas estão expressas em **milhões**. Na multiplicação, use as durações arredondadas para **1,67 h** e **1,50 h**.

### Gabarito — leitura inicial

1. **Títulos:** Guerreiras do K-Pop e Shrek.  
2. **Semanas:** encerradas em 7 e 14 de setembro de 2025.  
3. **Unidades:** visualizações e horas assistidas em milhões; duração em horas.

## 2. Organize as visualizações em matrizes

$A$ representa 7 set. 2025 e $B$ representa 14 set. 2025. As colunas são, nessa ordem, Guerreiras do K-Pop e Shrek.

```python
A = [[25.4, 3.6]]
B = [[22.6, 2.9]]

print("A =", A, "| ordem:", len(A), "x", len(A[0]))
print("B =", B, "| ordem:", len(B), "x", len(B[0]))
```

## 3. Consolide as duas semanas

$C=A+B$ por entradas correspondentes.

```python
C = [[
    A[0][0] + B[0][0],
    A[0][1] + B[0][1]
]]

print("C =", C)
print("Ordem de C:", len(C), "x", len(C[0]))
```

### Gabarito — interpretação de $C$

- $c_{1,1}=48,0$: visualizações acumuladas de Guerreiras do K-Pop.  
- $c_{1,2}=6,5$: visualizações acumuladas de Shrek.  
- A unidade permanece milhões de visualizações porque foram somados valores do mesmo indicador e da mesma unidade.

## 4. Relacione visualizações e duração

$D$ registra a duração de Guerreiras do K-Pop e Shrek, nessa ordem.

```python
D = [
    [1.67],
    [1.50]
]

print("D =", D)
print("Ordem de D:", len(D), "x", len(D[0]))
```

### Gabarito — compatibilidade

- $C$ tem ordem $1\times2$.  
- $D$ tem ordem $2\times1$.  
- As dimensões internas são iguais a 2.  
- Portanto, $H=C\cdot D$ é possível e terá ordem $1\times1$.

```python
linhas_C = len(C)
colunas_C = len(C[0])
linhas_D = len(D)
colunas_D = len(D[0])

print("Ordem de C:", linhas_C, "x", colunas_C)
print("Ordem de D:", linhas_D, "x", colunas_D)
print("Dimensões internas:", colunas_C, "e", linhas_D)
print("Compatíveis?", colunas_C == linhas_D)
print("Ordem esperada de H:", linhas_C, "x", colunas_D)
```

## 5. Calcule a estimativa de horas assistidas

$H=48,0\cdot1,67+6,5\cdot1,50=89,91\approx89,9$.

```python
H = [[
    C[0][0] * D[0][0] + C[0][1] * D[1][0]
]]

print("H =", H)
print("Estimativa arredondada:", round(H[0][0], 1), "milhões de horas")
```

## 6. Valide o resultado

A soma dos valores publicados é $42,4+5,4+37,7+4,4=89,9$. A estimativa obtida por multiplicação coincide com o total publicado quando arredondada para uma casa decimal.

```python
total_publicado = 42.4 + 5.4 + 37.7 + 4.4
estimativa = H[0][0]
diferenca = abs(total_publicado - estimativa)

print("Total publicado:", total_publicado, "milhões de horas")
print("Total estimado:", round(estimativa, 2), "milhões de horas")
print("Estimativa arredondada:", round(estimativa, 1), "milhões de horas")
print("Diferença antes do arredondamento:", round(diferenca, 2), "milhões de horas")
```

## 7. Gabarito — análise dos resultados

1. **Guerreiras do K-Pop**, com 48,0 milhões de visualizações.  
2. Aproximadamente **89,9 milhões de horas assistidas**.  
3. A multiplicação relacionou as visualizações acumuladas à duração de cada título, produzindo uma estimativa do tempo total assistido.  
4. A análise considera apenas dois títulos, duas semanas e dados globais do Top 10.

## 8. Exemplo de conclusão

Nas duas semanas, Guerreiras do K-Pop acumulou 48,0 milhões de visualizações, enquanto Shrek somou 6,5 milhões. Juntos, os títulos alcançaram aproximadamente 89,9 milhões de horas assistidas. A análise considera apenas dois filmes e não representa todo o catálogo nem todos os usuários da plataforma.
