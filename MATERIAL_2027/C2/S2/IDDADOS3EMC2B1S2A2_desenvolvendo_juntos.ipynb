## Contexto

Duas unidades escolares estimaram as participações nas oficinas **IA para estudar e pesquisar** e **IA para criar conteúdos**.

- **Linhas:** faixas etárias — 13 a 14 anos; 15 a 17 anos.
- **Colunas:** oficinas — estudo e pesquisa; criação de conteúdos.
- **Entradas:** participações estimadas.

Execute as células na sequência e acompanhe as orientações do professor.

## Etapa 1 — Representar U1 e U2

Crie as duas matrizes mantendo a mesma ordem de linhas e colunas.

**Código Python**

```python
U1 = [
    [62, 23],
    [75, 35]
]

U2 = [
    [55, 21],
    [82, 38]
]

print("U1 =", U1)
print("U2 =", U2)
```

## Etapa 2 — Conferir posições correspondentes

Observe alguns pares de entradas que ocupam a mesma posição nas duas matrizes.

**Código Python**

```python
print("Posição [0][0]:", U1[0][0], "e", U2[0][0])
print("Posição [0][1]:", U1[0][1], "e", U2[0][1])
print("Posição [1][0]:", U1[1][0], "e", U2[1][0])
print("Posição [1][1]:", U1[1][1], "e", U2[1][1])
```

## Etapa 3 — Construir a matriz T

Some as entradas correspondentes de U1 e U2 para reunir as estimativas das duas unidades.

**Código Python**

```python
T = [
    [U1[0][0] + U2[0][0], U1[0][1] + U2[0][1]],
    [U1[1][0] + U2[1][0], U1[1][1] + U2[1][1]]
]
```

## Etapa 4 — Visualizar a saída

Execute a célula e observe a matriz resultante.

**Código Python**

```python
T
```

## Etapa 5 — Comparar U1, U2 e T

Confira como cada entrada de T foi obtida a partir das posições correspondentes.

**Código Python**

```python
print("[0][0]:", U1[0][0], "+", U2[0][0], "=", T[0][0])
print("[0][1]:", U1[0][1], "+", U2[0][1], "=", T[0][1])
print("[1][0]:", U1[1][0], "+", U2[1][0], "=", T[1][0])
print("[1][1]:", U1[1][1], "+", U2[1][1], "=", T[1][1])
```

## Etapa 6 — Interpretar os resultados

Execute a consulta e registre no roteiro em Word o que cada valor representa no planejamento das oficinas.

**Código Python**

```python
print("T[1][0] =", T[1][0])
print("T[1][1] =", T[1][1])
```
