## Dados da atividade

**Código Python**


```python
T = [
    [117, 44],
    [157, 73]
]

R = [
    [8, 6],
    [5, 7]
]

C = [
    [10, 5],
    [20, 10]
]

print("T =", T)
print("R =", R)
print("C =", C)
```

## Etapa 1 — Atualizar as participações

A adição é feita por entradas correspondentes.


```python
P = [
    [T[0][0] + R[0][0], T[0][1] + R[0][1]],
    [T[1][0] + R[1][0], T[1][1] + R[1][1]]
]

P
```

**Saída esperada:** `[[125, 50], [162, 80]]`


## Etapa 2 — Recalcular o orçamento

Aplique o procedimento linha × coluna.


```python
o11_novo = P[0][0] * C[0][0] + P[0][1] * C[1][0]
o12_novo = P[0][0] * C[0][1] + P[0][1] * C[1][1]
o21_novo = P[1][0] * C[0][0] + P[1][1] * C[1][0]
o22_novo = P[1][0] * C[0][1] + P[1][1] * C[1][1]

O_atualizado = [
    [o11_novo, o12_novo],
    [o21_novo, o22_novo]
]

O_atualizado
```

**Saída esperada:** `[[2250, 1125], [3220, 1610]]`
