## 2. Organize as visualizações em matrizes

A matriz $A$ representa a semana encerrada em **7 de setembro de 2025**.  
A matriz $B$ representa a semana encerrada em **14 de setembro de 2025**.

As duas matrizes devem ter ordem $1\times2$, com as colunas nesta ordem: Guerreiras do K-Pop, Shrek.

```python
# Substitua os zeros pelos dados corretos da tabela.
A = [[0.0, 0.0]]
B = [[0.0, 0.0]]

print("A =", A)
print("Ordem de A:", len(A), "x", len(A[0]))
print("B =", B)
print("Ordem de B:", len(B), "x", len(B[0]))
```

## 3. Consolide as duas semanas

Calcule $C=A+B$ por entradas correspondentes. A matriz $C$ deve reunir as visualizações dos mesmos títulos nas duas semanas.

```python
C = [[
    A[0][0] + B[0][0],
    A[0][1] + B[0][1]
]]

print("C =", C)
print("Ordem de C:", len(C), "x", len(C[0]))
```

## 4. Relacione visualizações e duração

A matriz $D$ deve registrar a duração dos títulos, na mesma ordem usada nas colunas de $C$. Use **1,67 h** e **1,50 h**.

```python
# Substitua os zeros pelas durações corretas.
D = [
    [0.0],
    [0.0]
]

print("D =", D)
print("Ordem de D:", len(D), "x", len(D[0]))
```

### Verifique a compatibilidade

Antes de multiplicar, responda:

1. Qual é a ordem de $C$?  
2. Qual é a ordem de $D$?  
3. As dimensões internas são iguais?  
4. Qual será a ordem de $H=C\cdot D$?

Registre as respostas nesta célula.

```python
linhas_C = len(C)
colunas_C = len(C[0])
linhas_D = len(D)
colunas_D = len(D[0])

print("Ordem de C:", linhas_C, "x", colunas_C)
print("Ordem de D:", linhas_D, "x", colunas_D)
print("Dimensões internas:", colunas_C, "e", linhas_D)
print("Compatíveis?", colunas_C == linhas_D)
print("Ordem esperada de H:", linhas_C, "x", colunas_D)
```

## 5. Calcule a estimativa de horas assistidas

Calcule $H=C\cdot D$. O resultado deve representar o total estimado de horas assistidas pelos dois títulos nas duas semanas.

```python
H = [[
    C[0][0] * D[0][0] + C[0][1] * D[1][0]
]]

print("H =", H)
print("Estimativa arredondada:", round(H[0][0], 1), "milhões de horas")
```

## 6. Valide o resultado

Some as horas assistidas publicadas para os quatro registros do recorte e compare com $H$.

```python
total_publicado = 42.4 + 5.4 + 37.7 + 4.4
estimativa = H[0][0]
diferenca = abs(total_publicado - estimativa)

print("Total publicado:", total_publicado, "milhões de horas")
print("Total estimado por H:", round(estimativa, 1), "milhões de horas")
print("Diferença antes do arredondamento:", round(diferenca, 2), "milhões de horas")
```
