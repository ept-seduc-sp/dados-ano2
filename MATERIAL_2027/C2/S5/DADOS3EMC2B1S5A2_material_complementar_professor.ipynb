O objetivo é verificar previsões de um modelo linear com **duas variáveis de entrada** e interpretar como distância e tempo alteram o valor estimado.

**Modelo utilizado:** `V = 6,55 + 4,80d + 0,925t`

- `V`: valor estimado da corrida;
- `d`: distância, em km;
- `t`: tempo em tarifa horária, em minutos.

> **Recorte didático:** a tarifa horária oficial é R$ 55,50 por hora. Para esta atividade, foi convertida para R$ 0,925 por minuto. O modelo é uma simplificação didática e não reproduz todas as condições operacionais de uma corrida real.

**Uso docente:** as células estão executadas para permitir conferência imediata das saídas esperadas.

## Etapa 1 — Verifique as situações A, B, C e D

As situações abaixo são as mesmas do material complementar. Execute a célula e compare as saídas com os cálculos que você fez no documento.

```python
bandeirada = 6.55
tarifa_km = 4.80
tarifa_minuto = 0.925

situacoes = [
    ("A", 7, 0),
    ("B", 7, 6),
    ("C", 10, 0),
    ("D", 10, 6)
]

for nome, d, t in situacoes:
    valor = bandeirada + tarifa_km * d + tarifa_minuto * t
    print(f"Situação {nome}: {d} km, {t} min -> R$ {valor:.2f}".replace(".", ","))
```

## Etapa 2 — Compare uma variável de cada vez

Execute a célula e observe as diferenças entre **A e B** e entre **A e C**. Depois, registre a interpretação no documento complementar.

```python
valor_A = bandeirada + tarifa_km * 7 + tarifa_minuto * 0
valor_B = bandeirada + tarifa_km * 7 + tarifa_minuto * 6
valor_C = bandeirada + tarifa_km * 10 + tarifa_minuto * 0

diferenca_tempo = valor_B - valor_A
diferenca_distancia = valor_C - valor_A

print(f"A -> B (apenas o tempo muda): + R$ {diferenca_tempo:.2f}".replace(".", ","))
print(f"A -> C (apenas a distância muda): + R$ {diferenca_distancia:.2f}".replace(".", ","))
```

## Etapa 3 — Nova previsão

Calcule com Python a **Situação E**:

- distância: **12 km**;
- tempo em tarifa horária: **10 min**.

Execute a célula e registre o resultado no documento complementar.

```python
d = 12
t = 10
valor_E = bandeirada + tarifa_km * d + tarifa_minuto * t

print(f"Situação E: {d} km, {t} min -> R$ {valor_E:.2f}".replace(".", ","))
```

## Interpretação esperada

- **A → B:** a distância permanece em 7 km e o tempo aumenta 6 min; a previsão aumenta **R$ 5,55**.
- **A → C:** o tempo permanece em 0 min e a distância aumenta 3 km; a previsão aumenta **R$ 14,40**.
- **Situação E:** **R$ 73,40**.
- Mantida a distância, cada minuto adicional em tarifa horária acrescenta **R$ 0,925** ao valor estimado.
- Mantido o tempo, cada quilômetro adicional acrescenta **R$ 4,80** ao valor estimado.
- O modelo pode ser representado por um **plano** porque utiliza duas variáveis de entrada — distância e tempo — para estimar o valor da corrida.

**Mediação:** se os estudantes tentarem concluir que uma variável é sempre mais importante que a outra, retome que a comparação realizada considera acréscimos específicos (+3 km e +6 min). O foco é interpretar os coeficientes e isolar uma variável de cada vez.
