## Novas previsões com Python

**Modelo utilizado:** `V = 6,55 + 4,80d`

- `V`: valor estimado da corrida;
- `d`: distância percorrida, em km.

> **Recorte didático:** o modelo considera apenas a bandeirada de R$ 6,55 e a tarifa quilométrica de R$ 4,80 da Bandeira 1. Não representa todas as condições de uma corrida real.

## Etapa 1 — Execute as previsões definidas

As distâncias foram organizadas em quatro pares. Em cada par, a segunda distância é exatamente **1 km maior** que a primeira.

Execute a célula e registre os valores no documento complementar.

```python
bandeirada = 6.55
tarifa_km = 4.80

distancias = [8, 9, 12, 13, 17, 18, 23, 24]

for d in distancias:
    valor = bandeirada + tarifa_km * d
    print(f"{d} km -> R$ {valor:.2f}".replace(".", ","))
```

## Etapa 2 — Compare os pares

No documento complementar, compare: **8 e 9 km**, **12 e 13 km**, **17 e 18 km**, **23 e 24 km**.

Antes de executar a célula seguinte, registre sua conclusão sobre o que acontece com o valor estimado quando a distância aumenta **1 km**.

```python
pares = [(8, 9), (12, 13), (17, 18), (23, 24)]

for d1, d2 in pares:
    v1 = bandeirada + tarifa_km * d1
    v2 = bandeirada + tarifa_km * d2
    diferenca = v2 - v1
    print(f"{d1} -> {d2} km: diferença = R$ {diferenca:.2f}".replace(".", ","))
```
