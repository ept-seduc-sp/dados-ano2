## Previsões para uma operação de entregas

Use este notebook em conjunto com **DADOS3EMC2B1S5A3_AP3_professor.docx**. O objetivo é conferir os cálculos do roteiro e comparar um modelo com uma variável de entrada e outro com duas variáveis.

**Valores referenciais oficiais usados na atividade:**
- PVR da entrega principal: **R$ 26,22**;
- entrega adicional: **R$ 13,11**.

**Modelos:**
- `C1 = 26,22q`
- `C2 = 26,22q + 13,11a`

`q` = quantidade mensal de PVR da entrega principal  
`a` = quantidade mensal de entregas adicionais  

**Fonte:** Governo do Estado de São Paulo. Prestação de Serviços de Motofrete, data-base maio/2025.  
https://compras.sp.gov.br/wp-content/uploads/2025/12/P13_Estudo_SEM_Marcas_Revisao_Vol11_V1.pdf

## Etapa 1 — Confira as operações A e B

Os dados abaixo são os mesmos usados no roteiro. Execute a célula e compare as saídas com seus cálculos.

```python
valor_pvr = 26.22
valor_adicional = 13.11

operacoes = [
    ("A", 63, 42),
    ("B", 21, 34)
]

for nome, q, a in operacoes:
    custo_reta = valor_pvr * q
    custo_plano = valor_pvr * q + valor_adicional * a
    print(f"Operação {nome}: Modelo 1 = R$ {custo_reta:.2f} | Modelo 2 = R$ {custo_plano:.2f}".replace(".", ","))
```

## Etapa 2 — Compare a contribuição das entregas adicionais

Execute a célula e observe quanto a segunda variável acrescenta ao custo em cada operação.

```python
for nome, q, a in operacoes:
    parcela_adicionais = valor_adicional * a
    print(f"Operação {nome}: parcela das entregas adicionais = R$ {parcela_adicionais:.2f}".replace(".", ","))
```
